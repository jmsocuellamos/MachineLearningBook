## <font color='steelblue'> 1. ¿Qué son las tablas de frecuencias?


Las tablas de frecuencias son herramientas utilizadas en el análisis de datos para resumir la distribución de valores en una variable, o la relación entre las respuestas en varias variables. Estas tablas presentan información sobre el número de registros, o lo que es lo mismo, la frecuencia de cada valor observado o categoría de respuesta. Permite así visualizar y comprender mejor la distribución de los datos, esto es, cómo se distribuyen o reparten las respuestas, facilitando la identificación de patrones, tendencias, y relaciones cuando involucramos varias variables.

Las tablas de frecuencias para una variable categórica, cualitativa u ordinal, permite describir la prevalencia de cada categoría de respuesta sobre una muestra de sujetos de una población de interés. Si la respuesta es ordinal, permite identificar si existe alguna tendencia que esté relacionada con el orden de las respuestas.

Las tablas de frecuencias para dos variables permiten examinar la relación entre ellas. Estas tablas presentan la frecuencia de las diferentes combinaciones de respuesta (cruces) en todas las categorías de ambas variables. Permiten identificar las relaciones y asociaciones entre las variables, lo que puede ser útil para el análisis y la toma de decisiones.

Las tablas de frecuencias multivariantes se utilizan cuando se trabaja con más de dos variables, si bien tienen la misma utilidad de las tablas bivariantes: examinar las relaciones y asociaciones complejas entre las variables, para tener una visión más completa de como interaccionan entre sí.

Una vez tenemos claros estos conceptos teóricos, aterricemos para aprender a construirlas, interpretarlas y exportarlas para incluirlas en nuestros informes.


Las tablas de frecuencias son posiblemente el primer descriptivo que abordamos cuando nos encontramos con una base de datos. En la situación más sencilla, con una única variable categórica, dicha tabla se estructura en dos columnas:
- la primera enumera todas las respuestas/valores posibles
- la segunda contiene el número de veces que se repite cada valor en la base de datos que analizamos, esto es, la **frecuencia absoluta** ($F$).


Una vez obtenida la frecuencia absoluta, a partir de ella podemos calcular (y también representar):


* la **frecuencia relativa** ($f$), que es el cociente entre la frecuencia absoluta y el tamaño de la muestra;
* el **porcentaje de ocurrencia** asociado a cada respuesta,  que es la frecuencia relativa multiplicada por 100, esto es, expresada en porcentaje %;
* la **frecuencia absoluta/relativa acumulada**, de especial interés cuando la variable es de tipo ordinal, puesto que acumula la frecuencia de aparición por debajo de cada valor de respuesta.

Las tablas de frecuencias, como ya dijimos, se pueden obtener tanto para describir de modo univariante variables cualitativas (nominales y ordinales), como cuantitativas cuando se han discretizado o categorizado. Pero también permite describir relaciones entre varias variables. Veamos cómo conseguirlas.

## <font color='steelblue'> 2. Tablas de frecuencias para variables cualitativas nominales

Recordemos que las variables cualitativas nominales (a veces referidas como 'factores de clasificación') son aquellas con un número limitado de respuestas posibles, sin significado de tipo numérico u ordinal, como por ejemplo el color de los ojos o el color del pelo.

La tabla de frecuencias para este tipo de variables se calcula con las frecuencias absolutas y relativas (o porcentajes) de cada categoría o nivel de respuesta.

Para construir tablas de frecuencia podemos utilizar 3 métodos diferentes:

- **`value_counts()`**, de *Pandas*, que proporciona directamente los conteos o frecuencias absolutas de cada una de las respuestas distintas en la variable;
- **`groupby()`** y **`count()` y/o `agg()`**, cuando queremos agrupar una variable, no necesariamente categórica, en cada uno de los niveles de respuesta de otra nominal, y contar cuántos registros existen en cada uno;
- **`crosstab()`**, que es la forma más elegante y específica en *Pandas* para construir tablas de frecuencias.

Para facilitar el trabajo de la obtención de las tablas de frecuencias y completar con el resto de información (frecuencias relativas y porcentajes) vamos a definir dos funciones. La primera de ellas es la tabla estándar, mientras que  en la segunda podemos introducir el orden en que queremos que parezcan los diferentes niveles de la variable de interés.

Ambas funciones vienen explicadas a través de los comentarios pertinentes.

````
# FUNCIONES PARA GENERAR TABLAS DE FRECUENCIAS CON VARIABLES NOMINALES

#Función para generar una tabla de frecuencias sin establecer orden

def tfreq(df, varcol):
  '''
  Función para generar una tabla de frecuencias sin establecer orden

  Parámetros de entrada:
    - df: data frame
    - varcol: variable de interés sobre la que se desea obtener la tabla
            de frecuencias

  Resultados:
    Tabla de frecuencias con frecuencias absolutas, frecuencias relativas,
    y porcentajes.
  '''
  # Generamos tabla de frecuencias absolutas
  tabla = (df
          .groupby(varcol)
          .agg(Fa = (varcol, "count")) # frecuencia absoluta
          .reset_index())
  # Total de la tabla
  total = sum(tabla.Fa)
  # Frecuencias relativas
  tabla["fr"] = round(tabla["Fa"]/ total, 4)
  # porcentaje por filas
  tabla["Percent"] = round(100*tabla["fr"],2)
  return(tabla)

# Función para generar una tabla de frecuencias con orden preestablecido de
# las categorías

def tfreq_orden(df, varcol, cats_ordered):
  '''
  Función para generar una tabla de frecuencias estableciendo un orden
  para las categorías

  Parámetros de entrada:
    - df: data frame
    - varcol: variable de interés sobre la que se desea obtener la tabla
            de frecuencias
    - cats_ordered: Orden preestablecido de las categorías

  Resultados:
    Tabla de frecuencias con frecuencias absolutas, frecuencias relativas,
    y porcentajes.
  '''
  # Creamos un vector con el número de categorías
  cat_num = list(range(len(cats_ordered)))
  # Recodificamos a etiquetas numéricas
  for i in cat_num:
    df[varcol] = df[varcol].replace({cats_ordered[i]: str(cat_num[i])})
  # obtenemos la tabla de frecuencias
  tabla = tfreq(df, varcol)
  # Volvemos a las etiquetas originales
  for i in cat_num:
    tabla[varcol] = tabla[varcol].replace({str(cat_num[i]): cats_ordered[i]})
    df[varcol] = df[varcol].replace({str(cat_num[i]): cats_ordered[i]})
  # Devolvemos la tabla
  return(tabla)
````

## <font color='steelblue'> 3. Tablas de frecuencias para variables cualitativas ordinales

Recordemos que las variables cualitativas ordinales son aquellas que siendo categóricas, sus niveles siguen algún tipo de orden. Por ejemplo, es ordinal una variable que registra el grado de desarrollo de cierta enfermedad o el nivel de estudios de los sujetos.

La tabla de frecuencias para este tipo de variables se obtiene de igual forma que para las nominales, pero al contener información ordinal, ésta ha de ser tenida en cuenta para enriquecer el descriptivo, tanto a nivel numérico como gráfico.


Para estas variables es interesante obtener además de las frecuencias descritas anteriormente, sus valores acumulados: **frecuencia absoluta acumulada** y **frecuencia relativa acumulada**, respetando el orden de las respuestas para realizar el acumulado. Nos dan así información sobre el número y la proporción de datos que quedan por debajo (y por encima) de cada uno de los valores consecutivos de respuesta. Para obtener las tablas usamos la función `value_counts()` del paquete base de **Python**, y para mejorar la visualización *Pandas*.

Para ejemplificar utilizaremos las bases de datos `stroke` y `alcohol`.

### <font color='steelblue'> Análisis Numérico

Para facilitar el trabajo con este tipo variables definimos (de forma similar al caso anterior) funciones específicas para obtener las tablas de frecuencias en variables ordinales, modificando las que ya presentamos para variables nominales, de forma que podamos incorporar y representar también las frecuencias acumuladas:

* **tfreq_ordinal(df, varcol)**, cuando las categorías de la variable ya vienen ordenadas; generalmente cuando están codificadas mediante números.
* **tfreq_ordinal_ord(df, varcol, cats_ordered)**, cuando la variable viene codificada con etiquetas de texto o no en el orden lógico; 'cats_ordered' indicará el orden en que deseamos que aparezcan los niveles de la variable al obtener la tabla.

````
# FUNCIONES PARA OBTENER LAS TABLAS DE FRECUENCIAS DE VARIABLES ORDINALES

def tfreq_ordinal(df, varcol):
  '''
  Función para generar una tabla de frecuencias para variables ordinales donde
  las categorías ya están ordenadas.

  Parámetros de entrada:
    - df: data frame
    - varcol: variable de interés sobre la que se desea obtener la tabla
            de frecuencias

  Resultados:
    Tabla de frecuencias con frecuencias absolutas, frecuencias relativas,
    frecuencias acumuladas y porcentajes.
  '''
  # Tabla de frecuencias
  tabla = (df
          .groupby(varcol)
          .agg(Fa = (varcol, "count"))
          .reset_index())
  # Totales
  total = sum(tabla.Fa)
  # Frecuencias relativas
  tabla["fr"] = round(tabla["Fa"]/ total, 4)
  # Porcentajes
  tabla["Percen"] = round(100*tabla["fr"],2)
  # Frecuencia absoluta acumulada
  tabla["Facum"] = tabla["Fa"].cumsum()
  # Frecuencia relativa acumulada
  tabla["facum"] = round(tabla["Facum"]/ total,4)
  # Porcentajes acumulados
  tabla["Percenacum"] = round(100*tabla["facum"],2)
  return(tabla)

def tfreq_ordinal_ord(df, varcol, cats_ordered):
  '''
  Función para generar una tabla de frecuencias para variables ordinales donde
  se establece el orden de als categorías.

  Parámetros de entrada:
    - df: data frame
    - varcol: variable de interés sobre la que se desea obtener la tabla
            de frecuencias
    - cats_ordered: lista con el orden de las categorías que se desea utilizar

  Resultados:
    Tabla de frecuencias con frecuencias absolutas, frecuencias relativas,
    frecuencias acumuladas y porcentajes.
  '''
  # Codificamos según el orden establecido
  cat_num = list(range(len(cats_ordered)))
  for i in cat_num:
    df[varcol] = df[varcol].replace({cats_ordered[i]: str(cat_num[i])})
  tabla = tfreq_ordinal(df, varcol)
  # Volvemos a las etiquetas oroginales
  for i in cat_num:
    tabla[varcol] = tabla[varcol].replace({str(cat_num[i]): cats_ordered[i]})
    df[varcol] = df[varcol].replace({str(cat_num[i]): cats_ordered[i]})
  # Devolvemos la tabla
  return(tabla)
````

Obtenemos la tabla de frecuencias para la variable `smoking_status` del dataframe `stroke`. Dicha variable tiene niveles 'Unknown', 'formerly smoked', 'never smoked' y 'smokes'. Si aplicamos directamente la tabla de frecuencias  los porcentajes acumulados no tendrán sentido, al no calcularse mediante lo que sería el orden lógico de los niveles de la variable.

Veamos que ocure si utilizamos la función sin establecer el orden:

````
# Ejecutamos la función para obtener la tabla de frecuencias
tfreq_ordinal(stroke, "smoking_status")
````

````
# Seleccionamos el orden
cats_ordered = ["never smoked", "formerly smoked", "smokes", "Unknown"]
# Calculamos la tabla
tfreq_ordinal_ord(stroke, "smoking_status", cats_ordered)
````

La interpretación de los resultados es así más comprensible: los 'no fumadores' junto con los 'fumadores esporádicos' representan el 54.34% de los sujetos del estudio. Además hay un 30.22% de sujetos sobre los que no constan sus hábitos como fumadores.

## <font color='steelblue'> 4. Tablas de frecuencias para variables cuantitativas discretizadas

Para las variables cuantitativas, donde los datos son números, no tiene sentido obtener la tabla de frecuencias para cada uno de estos, dado que lo esperable es que no tengamos muchas repeticiones de un mismo valor de la variable. La tabla de frecuencias para una variable cuantitativa tiene sentido si previamente se ha discretizado la variable, esto es, si se ha categorizado agrupando los valores por intervalos, o incluso asignando un valor común (marca de clase) a todas las observaciones en un mismo intervalo. Conseguimos así una variable de tipo ordinal, y aplicamos las reglas anteriores para la construcción de tablas de frecuencias para variables ordinales.

Ya trabajamos la categorización de variables cuantitativas en cuadernos anteriores, con la función **`cut()`**.

````
# Renombramos variable 'PM2.5' para evitar el punto (.)
pollution.rename(columns={'PM2.5': 'PM2_5'}, inplace=True)
# Agrupamos en función de los niveles de cada variable y guardamos en una nueva variable
# Fijamos el valor cero para el extremo inferior y el máximo para el extremo superior
# Los intervalos se calculan de forma automática en función de los puntos frontera especificados
pollution['SO2_cat'] = pd.cut(pollution.SO2, [0, 0.02, 0.05, 0.15, 1, pollution.SO2.max()], include_lowest = True, right = False)
pollution['NO2_cat'] = pd.cut(pollution.NO2, [0, 0.03, 0.06, 0.2, 2, pollution.NO2.max()], include_lowest = True, right = False)
pollution['O3_cat'] = pd.cut(pollution.O3, [0, 2, 9, 15, 50, pollution.O3.max()], include_lowest = True, right = False)
pollution['CO_cat'] = pd.cut(pollution.CO, [0, 0.03, 0.09, 0.15, 0.5, pollution.CO.max()], include_lowest = True, right = False)
pollution['PM10_cat'] = pd.cut(pollution.PM10, [0, 30, 80, 150, 600, pollution.PM10.max()], include_lowest = True, right = False)
pollution['PM2_5_cat'] = pd.cut(pollution.PM2_5, [0, 15, 35, 75, 500, pollution.PM2_5.max()], include_lowest = True, right = False)
````

## <font color='steelblue'> 5. Tablas de frecuencias para dos variables cualitativas

Una vez hemos estudiado de forma individualizada cómo se distribuyen las variables en una base de datos, el siguiente paso suele ser el de buscar relaciones y asociaciones entre variables distintas.

Cuando tenemos dos variables categóricas (o categorizadas), podemos cruzar todos los niveles de respuesta de las dos, y contar cuántos datos hay en cada cruce. Esto lo reflejamos en una tabla de frecuencias bivariada que se denomina **tabla de contingencia**. En sus filas aparecerán  todos los niveles de respuesta de una de las variables, y en las columnas los correspondientes para la otra;  en las celdas se dispone el número de observaciones que están clasificados en cada cruce de respuestas. En ocasiones nos resultará más interesante calcular los porcentajes asociados a dichos conteos, que podemos calcular de varios modos en función de nuestros objetivos:

* **Porcentajes globales**, calculados respecto al total de casos analizados, y que tienen interés para comprobar si todos los cruces tienen similares proporciones de respuesta o existe alguna tendencia hacia ciertos niveles.
* **Porcentajes por filas**, respecto al total de casos de cada fila. Interesa entonces investigar si el reparto de respuestas para la variable columna en cada una de las respuestas filas es similar o no.
* **Porcentajes por columnas**, respecto al total de casos de cada columna. Se investiga si el reparto de respuestas para la variable fila en cada una de las respuestas columnas es similar o no.

Construimos las tablas de contingencia con la función `crosstab()` de Pandas, cuyos argumentos principales trabajaremos a continuación:
```
crosstab(index, columns, values=None,
           rownames=None,colnames=None,
           margins=False, margins_name='All',
           dropna=True,
           normalize=False)
```

El primer argumento, `index`, es la variable fila, y el segundo, `columns`, es la variable columna. Por defecto los valores en las celdas son los conteos, si bien se pueden calcular las proporciones, con el argumento `normalize=True`, e incluso los porcentajes, multiplicando esta última tabla resultante por 100.

Para ejemplificar todos los cálculos y gráficos utilizaremos los bancos de datos `stroke` y `pollution`.

### <font color='steelblue'> Análisis Numérico

Para mostrar cómo se construyen e interpretan las tablas de contingencia, utilizamos las variables `stroke` y `gender` en la base de datos 'stroke'. En primer recodificamos la variable `stroke` para asignar etiquetas más comprensibles.

````
# Recodificamos
stroke['stroke'] = stroke['stroke'].replace({0: "Non stroke", 1: "Stroke"})
# Creamos la tabla de frecuencias
# tabla de contingencia para sexo y provincia: conteos
pd.crosstab(stroke["stroke"], stroke["gender"])
````

Obtenemos la tabla de porcentajes globales, es decir, la suma de todos los porcentajes es igual a 100.

````
# tabla de contingencia para sexo y provincia: proporciones
round(pd.crosstab(stroke["stroke"], stroke["gender"],normalize=True)*100, 2)
````

Los porcentajes globales muestran los repartos respecto del total: en la muestra el porcentaje de mujeres con accidentes cerebro-vasculares fue de 2.76% y el de hombres de 2.11%: bastante similares. Si queremos completar la tabla con el total de hombres y mujeres, e incluso con el total de observaciones en la base de datos, lo indicamos con el argumento **`margins=True`**.


````
# tabla de contingencia para sexo y provincia: proporciones
round(pd.crosstab(stroke["stroke"], stroke["gender"],normalize=True, margins=True)*100, 2)
````

Si queremos valorar el reparto por sexos entre los sujetos que sufrieron un accidente, acudimos a los **porcentajes fila**. Si queremos valorar la incidencia de accidentes cerebrovaculares en las mujeres y en los hombres, hemos de acudir a los **porcentajes columna**.

````
# Tabla de contingencia con porcentajes relativizados por filas
round(pd.crosstab(stroke["stroke"], stroke["gender"], normalize="index", margins=True, margins_name="TOTAL")*100, 2)
````

Podemos ver que las mujeres representan el 56.63% de todos los sujetos que han sufrido un incidente cerebrovascular, siendo hombres el 43.37% restante. De entre los que sufren un incidente cerebrovascular parece observarse una mayor incidencia entre las mujeres.

````
# Tabla de contingencia con porcentajes relativizados por columnas
round(pd.crosstab(stroke["stroke"], stroke["gender"], normalize="columns", margins=True, margins_name="TOTAL")*100, 2)
````

La incidencia de accidentes cerebrovasculares en hombres (5.11%) no difiere mucho de la incidencia en mujeres (4.71%).

## <font color='steelblue'> 6. Tablas de frecuencias para múltiples variables cuantitativas

Las tablas de contingencia multivariantes son aquellas en las que se muestra información sobre más de 2 variables categóricas. No es recomendable introducir más de 4 o 5 variables por la confusión que pueden generar.

Nos centraremos en las tablas de contingencia con 3 variables categóricas, utilizaremos dos de ellas como índices, o variables fila, agrupadas en una lista y en el primer argumento de la función, y la tercera en las columnas, ubicada en el segundo argumento de la función `crosstab()`.

Para mostrar como obtener las tablas de frecuencias vamos a utilizar como ejemplo la base de datos `stroke`, donde queremos estudiar la tabla  correspondiente a las variables `stroke`, `work_type`, y `smoking_status`, es decir, queremos como pueden influir el tipo de trabajo y la adicción al tabaco a la hora de sufrir una ataque al corazón o no.

````
# Tabla de frecuencias (absolutas) trivariante
pd.crosstab([stroke["stroke"] , stroke["smoking_status"]], stroke["work_type"])
````

Podemos obtener los porcentajes asociados que nos resultaran más fáciles de interpretar.

````
# Tabla de frecuencias (absolutas) trivariante
round(pd.crosstab([stroke["stroke"] , stroke["smoking_status"]], stroke["work_type"],
                  normalize=True, margins=True, margins_name="Subtotal")*100, 2)
````

¿Cómo interpretamos los resultados obtenidos?

Si deseamos calcular los porcentajes por filas normalizamos por el índice:

````
# Tabla de frecuencias (absolutas) trivariante
round(pd.crosstab([stroke["stroke"] , stroke["smoking_status"]], stroke["work_type"],
                  normalize="index", margins=True, margins_name="Subtotal")*100, 2)
````

En este caso podemos ver con los que trabajan en el secto privado son los que presentan mayores porcentajes de incidencia para las combinaciones de `stroke` y `smoking_status`.

Para los que no han sufrido un ataque al corazón no fumar nos da los porcentajes más altos independeintemente del tipo de trabajo. ## <font color='steelblue'> 7. Exportar las tablas de frecuencias

De esta manera, se garantiza la disponibilidad de los resultados del análisis en un formato que pueda ser compartido, utilizado en informes o integrado en otras herramientas de análisis de datos.

Formatos habituales de informes son $\LaTeX$, *Markdown*, _HTML_, o el que proporcionan los ficheros *Word*. Existen algunas funciones en *Pandas* que aplicadas a tablas de frecuencias nos permite exportarlas a estos formatos, que podemos copiar y pegar para llevarlos a los documentos/informes de referencia que estamos construyendo, y que son:
- **`to_string()`**, que representa la tabla en texto plano
- **`to_markdown()`**, proporciona la sintaxis para representar la tabla en Markdwon
- **`to_latex()`**, proporciona la sintaxis para representar la tabla en $\LaTeX$
- **`to_html()`**, proporciona la sintaxis para representar la tabla en html.

Para ampliar conocimientos sobre estas funciones se recomienda consultar los siguientes enlaces: [to_string()](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.to_string.html), [to_markdown()](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.to_markdown.html), [to_html()](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.to_html.html) y [to_latex()](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.to_latex.html).
