## <font color='steelblue'> Elementos básicos

Puesto que ya controlamos la jerga básica estadística, repasemos la jerga específica que utilizaremos cuando emprendemos la modelización estadística.

* El modo óptimo de responder cuestiones de investigación o resolver problemas de interés pasa por plantear un diseño de experimentos. Sin embargo, como ya comentamos, no siempre será el caso, y en ocasiones utilizaremos datos secundarios que ya otros recopilaron y quizá con otros fines distintos a los nuestros. Este hecho no anula la fiabilidad de las conclusiones que se deriven con un modelo estadística, pero las limita por la representatividad de la muestra.

* Hablaremos de **muestra** como el conjunto de todos los datos recopilados sobre las variables de interés, para dar respuesta a las cuestiones planteadas a través de algún modelo estadístico. En la bibliografía también encontrarás algunas fuentes que denominan muestras a las observaciones individualizadas sobre todas las variables utilizadas.

* En modelización estadística habitualmente tenemos una característica o variable objetivo que tratamos de relacionar con otras que hemos observado, porque nos interesa especialmente predecirla en función de ellas. Se trata de la **variable respuesta**, también reconocida como **target** en los modelos de aprendizaje. En la bibliografía más tradicional se la reconoce como variable dependiente, por su dependencia de las que actuarán como predictoras. Si la respuesta es de tipo numérico utilizaremos modelos de regresión para buscar relaciones con el resto de variables. Si es de tipo categórico utilizaremos modelos de clasificación, pues el objetivo será clasificar a los sujetos respecto de tal variable, también denominada etiqueta en la jerga de aprendizaje automático.

* **Variables predictoras o explicativas** serán pues el conjunto de atributos o características disponibles sobre cada uno de los datos o registros de la muestra y con los que pretendemos predecir la variable respuesta. A menudo se representan, para cada registro, con un vector que contiene información sobre las características observadas. Estas variables pueden ser de tipo numérico o categórico y recibirán tratos distintos en los modelos que estudiaremos. En la jerga del Machine Learning, las variables predictoras se denominan **características**.

* Aludiremos con el término **parámetros** a aquellos descriptores de la población que pretendemos estimar, como por ejemplo la estatura media, o el tiempo medio de degradación, pero también la severidad de la relación lineal entre un predictor y una respuesta. Los modelos estadísticos plantearán fórmulas para relacionar los predictores con la respuesta a través de parámetros, y aproximarlos a través de los datos recopilados en la muestra.

* Surge después otro término que nos permite reajustar los modelos para conseguir mejores predicciones. Se trata de los **hiperparámetros** o parámetros libres vinculados a algunos modelos y que nos permiten plantear realmente modelos alternativos para las mismas variables y una misma formulación relacional.
Todos los algoritmos de ajuste de modelos utilizan unos valores por defecto para los hiperparámetros, si bien podemos manipularlos.

* Un concepto crítico en modelización tiene que ver con el criterio que se sigue para ajustar un modelo. El ajuste de un modelo está siempre basado en minimizar una **función de pérdida**, que mide la diferencia entre el valor de la respuesta predicha por el modelo estadístico y el valor observado.


* Finalmente, comentar que cualquier banco de datos con el que emprenderemos la modelización tendrá una **estructura matricial**, donde las filas identificarán los registros o sujetos disponibles, y las columnas representarán las variables medidas y a relacionar. El cruce de una fila y una columna contendrá el valor observado para el sujeto registrado en dicha fila, en la variable representada en la columna en cuestión.


La modelización estadística pretendrá buscar respuestas a objetivos o hipótesis sobre la caracterización de una población, aproximando  la relación entre las características implicadas e identificando qué condiciones generan respuestas distintas. El analista siempre buscará el mejor modelo que dé respuestas válidas a los objetivos planteados. Y ahí... el abanico de posibilidades es inmenso, como veremos.

## <font color="steelblue">Formulación</font>

Un modelo estadístico es realmente una fórmula matemática que nos va a permitir relacionar la variable respuesta con las predictoras, para predecir la primera a través de las segundas. A esta fórmula matemática determinista, llamada **componente sistemática** del modelo, se le añade el error, que representa la variabilidad intrínseca en la población (sujetos con las mismas caracterizaciones pueden no compartir una misma respuesta), la que no es explicable con la fórmula (que es una aproximación a la relación real), y el error en la medición (que es esperable que sea pequeño si el sistema de medición ha sido validado). Este error o desviación se conoce como **componente aleatoria**. Comencemos a expresar estas ideas matemáticamente.

Hablamos ya de variables predictoras, que vamos a identificar como $(X_1, X_2,X_3, …)$ y de una (o más) variable respuesta o *target* que identificaremos como $Y$, y cuya predicción denotaremos por $\hat{Y}$. La modelización estadística busca relacionar la predicción con las predictoras mediante una fórmula o modelo matemático que expresaremos con la componente sistemática:

\begin{equation}
  \hat{Y} = f(X_1,X_2,...,X_k)
\end{equation}

Si nuestro modelo es 'bueno', esta función debe reflejar el comportamiento medio esperado de la variable respuesta, esto es, aproximar bien $E(Y)$ con $\hat{Y}$. La desviación entre ambas la denotaremos por $\epsilon$, de modo que el modelo estadístico lo podemos formular como:

\begin{equation}
  Y = \hat{Y} + \epsilon = f(X_1,X_2,...,X_k) + \epsilon
\end{equation}

Por último, mencionar que el error $\epsilon$ en ocasiones puede tener una estructura más compleja que surgirá en modelos avanzados que estudiaremos también en este curso.


## <font color="steelblue">Tipos de modelos</font>

* **Modelos de reregresión lineal**, cuando el target es de tipo continuo.
* **Modelos de reregresión logística**, cuando el target es de tipo discreto.

### Evaluar y validar el modelo

Cuando se ha seleccionado un modelo que funciona, es habitual evaluarlo para valorar su rendimiento final en términos de capacidad explicativa, es decir, cómo son de buenas las características para expicar el comportamiento de la respuesta. En función del tipo de target (numérico o categórico) podemos definir diferentes métricas de validación del modelo. En este punto presentamos las más generales pero profundizaremos más en este tema conforme vayamos presentando los diferentes tipos de modelos.

Las métricas de evaluación se utilizan para medir el rendimiento de los modelos de aprendizaje automático. Al principio de esta introducción comentamos que la mayoría de los problemas se refieren a cuestiones resolubles por regresión o por clasificación; en función del objetivo y el tipo de respuesta, la evaluación del modelo será diferente. Veamos las alternativas.

#### Métricas para problemas de regresión

En las tareas de regresión el objetivo es predecir el valor numérico de la variable respuesta. Así pues, las métricas que surgen de manera natural para evaluar estos modelos son las basadas en distancias entre las observaciones y las predicciones.

La diferencia entre el valor real de la respuesta ($y$) y el valor predicho ($\hat{y}$) se llama **error de predicción**:

$$Error = y - \hat{y}$$

El cuadrado del error de predicción sobre los datos observados $\{y_1,y_2,..., y_n\}$, se llama **error cuadrático medio** (MSE o *Mean Squared Error*, en inglés), y se calcula como:

$$MSE = \frac{\sum_{i=1}^n (y_i - \hat{y}_i)^2}{n}$$

Otra métrica común es la **raíz cuadrada del error cuadrático medio** (RMSE). El RMSE es la métrica de regresión más utilizada.

$$RMSE = \sqrt{\frac{\sum_{i=1}^n (y_i - \hat{y}_i)^2}{n}}$$

Hay veces que se trabaja con conjuntos de datos que contienen valores atípicos. Una métrica adecuada para este tipo de conjuntos de datos es el **error medio absoluto** (MAE, *Mean Absolute Error*) que se calcula como:

$$MAE = \frac{\sum_{i=1}^n |y_i - \hat{y}_i|}{n}$$

Por último, tenemos el **error de porcentaje medio absoluto** (MAPE) que es una medida relativa que escala esencialmente la MAE para que se muestre en unidades de porcentaje en lugar de en unidades de la variable. Se calcula como:

$$MAPE = \frac{1}{n} \sum_{i=1}^n \left|\frac{y_i - \hat{y}_i}{y_i}\right|$$

#### Métricas para problemas de clasificación

En los problemas de clasificación el objetivo es predecir correctamente la categoría de clasificación de cada observación. Así pues, las métricas naturales para evaluar estos modelos estarán basadas en contabilizar las coincidencias entre la clasificación correcta y la conseguida o predicha con el modelo.

Para definir estos indicadores, partimos de identificar las clasificaciones posibles, cuestión que salvamos con una **matriz de confusión**, esto es, una tabla que muestra el número de predicciones correctas e incorrectas realizadas por un clasificador en todas las clases disponibles. En el caso de una respuesta binaria, 0=Negativo y 1=Positivo en relación al hecho de que se de determinada característica, las clasificaciones/predicciones posibles sobre el total de datos o registros disponibles son las que se muestran en la siguiente matriz:

|                     |       | **Valor** | **Predicho** |
|---------------------|-------|--------------------|--------------|
|                     |       |        **0**       |     **1**    |
| **Valor** | **0** |         TN         |      FP      |
| **Observado**       | **1** |         FN         |      TP      |

Se identifican pues, los siguientes números:



|Abrev.|Índice |Descripción|
|-----|-------------------------|----------------|
**TP**| *True Positive* o Verdaderos positivos|# positivos clasificados positivos (1,1)|
**FP**| *False Positive* o Falsos positivos|# negativos clasificados positivos (0,1)|
**TN**|*True Negative* o Verdaderos negativos|# negativos clasificados negativos (0,0)|
**FN**|*False Negative* o Falsos negativos|# positivos clasificados negativos (1,0)|


En base a estos números se definen las siguientes métricas de evaluación de los modelos de aprendizaje basados en clasificación:

- La **exactitud** (representado en la literatura por A, de *accuracy*) es la métrica más utilizada y muestra la capacidad del modelo para hacer predicciones correctas. Se calcula a partir de la proporción de observaciones clasificadas correctamente (TP + TN sobre el total), esto es,

$$Exactitud (A)= \frac{TP + TN}{TP + TN + FP + FN}.$$

>La exactitud es un indicador de la bondad del modelo para hacer predicciones correctas, por lo que sólo será útil cuando el conjunto de datos esté repartido de forma equilibrada entre las categorías de respuesta, o no nos interese la clasificación en una categoría concreta.

>Supongamos que hemos construido un clasificador de imágenes que distingue entre caballos y humanos, con 250 imágenes de caballos y otras 250 de humanos. Si el modelo predice correctamente 400 imágenes en total (caballos y humanos), su exactitud es de 0.8, o lo que es lo mismo, del 80% en términos porcentuales.


>Cuando tenemos un conjunto de datos sesgado o desequilibrado hacia alguna categoría, o simplemente nos importa más predecir una categoría (como podría ser el diagnóstico de un tumor maligno), necesitamos una perspectiva diferente sobre cómo evaluar el modelo. Por ejemplo, si tenemos 450 imágenes de caballos y 50 imágenes de humanos, hay una probabilidad del 90% (450/500) de que el caballo se prediga correctamente porque el conjunto de datos está dominado por los caballos. Y aunque la predicción sea nefasta para los humanos, la exactitud seguirá siendo buena.

>Surgen pues otras métricas que, en estas circunstancias, serán más útiles que la exactitud, como la precisión, el recuerdo y la puntuación F1.

- La **precisión** (representado habitualmente como P, de *precision*) representa el porcentaje de observaciones correctamente clasificadas, de entre las que se han clasificado como positivas.

$$Precisión (P)= \frac{TP}{TP + FP}.$$

>Por ejemplo, nos podría interesar especialmente saber, de las imágenes que se clasificaron como humanos, cuántas eran realmente de humanos.

- Por otro lado, el **recuerdo** (representado por R, de *recall*) muestra el porcentaje de las observaciones que siendo positivas se clasificaron correctamente, y nos resulta útil cuando nos interesa especialmente la clasificación correcta de una determinada clase.

$$Recuerdo (R)= \frac{TP}{TP+FN}.$$

>En el ejemplo de caballos y humanos, descubriríamos, con el recuerdo, de las imágenes que eran de humanos, qué porcentaje fueron correctamente clasificadas como humanos.

>Dado que existe cierta relación entre la precisión y la recuperación, y a menudo el aumento de una conlleva la disminución de la otra, se propuso otra métrica construida a partir de las dos, para conjugar las ventajas de ambas.

- La **puntuación F1 (o *F1 Score*)** es la media armónica de la precisión y la recuperación, y muestra lo bueno que es el modelo a la hora de clasificar todas las clases sin tener que equilibrar la precisión y la recuperación. Si la precisión o la recuperación son muy bajas, la puntuación F1 también lo será.

$$F1 = \frac{2 \cdot P \cdot R}{P + R}.$$

**Evaluación y validación: procedimiento**

Todas estas métricas, en problemas de regresión o clasificación, se evalúan sobre los datos utilizados en el ajuste (muestra de entrenamiento), para comparar las predicciones con las observaciones. Si además se ha dejado una muestra test fuera del entrenamiento, también esta se evalúa para juzgar la calidad del modelo a la hora de predecir observaciones que no ha utilizado en el ajuste. Con las métricas disponibles, se decide sobre la bondad del ajuste.

La validación del modelo consistirá en estudiar si este es estable, es decir, si no se ve afectado en demasía por la muestra de entrenamiento y se resuelve habitualmente con técnicas de **validación cruzada**. Esta técnica se basa en que se separan los datos en varios subconjuntos (de tamaños similares) que se van utilizando para testar la bondad del modelo cuando se entrena el modelo con el resto. Se obtiene una medida media de calidad, promediando las métricas obtenidas en los subconjuntos de validación, y una medida de variabilidad/estabilidad con sus desviaciones típicas y rango de variación.

# <font color="steelblue">4. Retos en la modelización estadística</font>

Al igual que cualquier otra tecnología, la construción de modelos estadísticos ofrece desafíos, relacionados tanto con la mejora de los datos como con la mejora de los modelos de ajuste.

No puede haber buenos modelos con malos datos, pero es un hecho que en el mundo real es raro encontrar "buenos datos" a la primera. Por lo general los datos presentan desorden, errores, incoherencias, y se convierte en todo un arte la labor de procesarlos (depurar, etiquetar, e incluso crear nuevas variables) para conseguir una base de datos apta para iniciar el ajuste de un modelo.
Por otro lado, una vez disponemos de una base de datos razonablemente buena, el ajuste del modelo no es una labor trivial. En principio se inicia la labor de elegir qué modelo o modelos son razonables para los objetivos y el tipo de datos. A continuación hay que ajustarlos y compararlos para tomar una decisión.

Ya en el ajuste se nos pueden presentar dos tipos de problemas: el infra-ajuste (*underfitting*) y el sobre-ajuste (*overfitting*). Veamos en qué consisten.

El **infra-ajuste** o infra-adaptación del modelo se produce cuando el modelo da malos resultados (en términos de métricas de error) con los datos de entrenamiento (por ejemplo, una clasificación correcta de sólo el 50% de la muestra). Un problema de infra-ajuste ocasiona un problema de sesgo en la predicción. Esto puede ser causado porque el modelo es demasiado simple para los datos de entrenamiento o los datos no contienen las características que se están tratando de predecir. Para reducir este tipo de problema se puede optar por:

* Ajustar modelos más complejos. Modelos más sencillos, como los modelos lineales, pueden dar resultados más simples que los más complejos, como los modelos aditivos.

* Añadir más datos de entrenamiento y seleccionar mejor las mejores características para predecir,o incluso proponer transformaciones de las variables.

* Aumentar la complejidad de los modelos,lo que se conoce como "reducir el nivel de regularización del modelo". Por ejemplo, aumentar la sensibilidad de los modelos aditivos para adaptar mejor las variaciones en el rango de variación de la respuesta.

El **sobre-ajuste** o sobre-adaptación se produce cuando un modelo funciona muy bien con los datos de entrenamiento, pero no con datos nuevos que el modelo no utilizó para ajustar. Esto repercute en un inflado de la varianza de la muestra de validación.


El sobre-ajuste se puede deber a que el modelo es demasiado complejo para los datos disponibles o el volumen de la muestra de entrenamiento es demasiado pequeño. Para aliviar este tipo de problema, podemos intentar:

* Usar modelos más sencillos, simplificar el modelo actual a través de los hiperparámetros de los que dependen, reducir el número de predictores incluidos, etc.

* Utilizar más datos de entrenamiento para el ajuste.

* Aumentar el nivel de regularización del modelo, o lo que es lo mismo, reducir la complejidad del modelo, penalizando a los modelos que generan un error excesivamente pequeño para la muestra de entrenamiento.

## <font color="steelblue">Paquetes patsy/Formulaic

El paquete de Python [`patsy`](https://pypi.org/project/patsy/) es útil para construir matrices de diseño en los modelos lineales. En 2021 dejó de ser desarrollado, y se recomienda la migración a [Formulaic](https://github.com/matthewwardrop/formulaic).

Por ejemplo, si tenemos alguna variable `y` y queremos hacer una regresión con otras variables $x1$, $x2$, y la interacción entre ellas, entonces simplemente escribimos:

```
import patsy
patsy.dmatrices("y ~ x1 + x2 + x1:x2", datos)
```

y `patsy` se encarga de construir las matrices adecuadas.


La versión de esta funcionalidad en `Formulaic` es también sencilla de implementar con:
```
!pip install formulaic
import formulaic
from formulaic import model_matrix
model_matrix('y ~x1 + x2 + x1:x2', datos)
```

Además estos paquetes:

* Permite especificar transformaciones de datos usando código Python arbitrario: en lugar de x, podríamos haber escrito log(x), (x > 0), o incluso log(x) si x > 1e-5, de lo contrario log(1e-5),
* Proporciona una gama de opciones convenientes para codificar variables categóricas, incluida la detección automática y la eliminación de redundancias.
* Sabe cómo aplicar "la misma" transformación utilizada en datos originales a datos nuevos, incluso para transformaciones complicadas como el centrado o la estandarización (crítico si desea utilizar su modelo para hacer predicciones),
* Tiene un modo incremental para manejar conjuntos de datos que son demasiado grandes para caber en la memoria al mismo tiempo.
* Proporciona un lenguaje para la especificación simbólica y legible por humanos de matrices de restricciones lineales.
* Cuenta con una API simple para la integración en paquetes estadísticos.
