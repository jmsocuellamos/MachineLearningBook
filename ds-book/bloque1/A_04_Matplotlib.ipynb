### <font color='steelblue'>Crear y visualizar un gráfico

Hay dos funciones básicas para la representación de gráficos: `plot()` y `show()`. Para utilizarlas, tendremos que utilizar el prefijo `plt.`

La **función [`plt.plot()](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html#matplotlib-pyplot-plot)`** recibe tres argumentos: los valores del eje $X$, los valores del eje $Y$, cada uno en formato lista, y el tercer argumento (opcional) con el color y tipo de línea del gráfico.

La **función `plt.show()`** se utiliza  al final, después de definir el gráfico, para visualizarlo sin mensajes accesorios. También se puede sustituir por `;` tras la última ejecución vinculada a `plt.plot()`.

En el ejemplo a continuación representamos una parábola. Para ello introducimos una secuencia de valores $x$, y para cada uno de ellos calculamos su cuadrado, $y=x^2$. Guardamos ambas secuencias en sendas listas.

````python
# valores de "x"
x = list(range(6))

# creamos "y" vacía y la rellenamos con los cuadrados de x
y = []
for i in x:
  # vamos añadiendo cada valor de "x" al cuadrado
  y.append(i**2)

# visualizamos ambas listas
print(x)
print(y)
````

Sin preocuparnos sobre especificaciones de estilo, creamos el gráfico solo con las listas 'x' e 'y'. Aparecen las especificaciones básicas: una línea continua azul que une los puntos introducidos $(x,y)$, sin marcadores para los puntos.

````python
plt.plot(x, y)
plt.show()
````

Fíjate, que si no utilizas `plt.show()` aparece un mensaje indicando la lista de objeto 'Line2D' que devuelve `pyplot.plot`, por lo que se recomienda escribir siempre `plt.show()` para mejor legibilidad de la salida.

````python
plt.plot(x, y)
````

Y que evitamos de nuevo con el `;`:

````python
plt.plot(x,y);
````

## <font color='steelblue'> 2. Especificaciones básicas


Cuando queremos configurar los estilos de un gráfico, podemos hacerlo de dos modos:

1. utilizar el tercer argumento en la función `plot()` para introducir, en *formato corto* las especificaciones entrecomilladas y en un orden concreto;
1. utilizar los argumentos correspondientes de la función `plot()` en *formato largo*,  y en el orden que se desee.

### <font color='steelblue'>Formato corto para líneas y marcadores

Cuando utilizamos el formato corto, incluimos, básicamente, un carácter para reconocer cada una de las siguientes características, y siempre en el mismo orden: ***color, tipo de marcador y tipo de línea***.


Así, si especificamos en formato corto como tercer argumento `'g^--'`, estaremos pidiendo un gráfico de color verde ('g' de *green*), con triángulos como símbolos para los puntos ('^') y con líneas discontinuas (`'--'`). Veamos cómo queda el gráfico anterior con estas especificaciones.

````python
plt.plot(x, y, 'g^--')
plt.show()
````

Las posibilidades para las características básicas son:
- **COLOR**. Se suele utilizar la inicial del color en inglés. Por ejemplo: 'r' hace referencia a rojo (*red*), 'b' a azul (*blue*), 'y' a amarillo (*yellow*). El color negro se identifica por 'k'. Si no se especifica nada, por defecto los puntos y líneas se dibujan en azul.

- **MARCADORES (o PUNTOS)**. Un marcador redondo se representa con 'o', un triángulo con '^', 's' si queremos un cuadrado (*square*), '*' si queremos un asterisco, 'd' para un diamante y 'D' para rombos.  Si no especificamos nada, por defecto no se pintan los puntos.

- **LÍNEAS**. Tenemos la opción '-' para líneas continuas, '--' para discontinuas, ':' para líneas punteadas, y '-.' para líneas con líneas y puntos.

Si no se especifica nada para las líneas, pero sí para los marcadores, aparecen solo los marcadores, sin líneas, salvo cuando el color es negro ('k'). En cambio, si solo definimos el color, aparecerá una línea continua sin marcadores.

Veamos a continuación algunos ejemplos más.

````python
# gráfico en negro, marcadores con círculos y línea punteada uniendo puntos
plt.plot(x, y, 'ko:')
plt.show()
````

````python
# gráfico en rojo, marcadores con diamantes y línea con puntos y rayas
plt.plot(x, y, 'rd-.')
plt.show()
````

````python
# solo marcadores (puntos) en amarillo y con forma de cuadrado
plt.plot(x, y, 'ys')
plt.show()
````

### <font color='steelblue'>Formato largo para líneas y marcadores

La función `plt.plot()` dispone de argumentos específicos para controlar las tres características configurables con el formato corto, así como algunas otras. En concreto, el color, el tipo de marcador para los puntos y el tipo de línea lo especificamos, en formato largo, con los **argumentos `color`, `marker` y `linestyle`**.

Además, algunos de los argumentos de `plot()` tienen un nombre largo -como `color`-, y una abreviatura -en este caso `c`-, que indicamos a continuación al describirlos:

* **`color/c`**, donde especificamos, entrecomillado, el nombre del color en inglés (`'green'`, `'red'`, `'black'`, `'yellow'`, `'purple'`, `'grey'`, etc.), o bien el *código hexadecimal* inicializado por una almohadilla (por ejemplo, `color='#bc2a30'`); en [color-hex.com](https://www.color-hex.com/color-names.html) puedes encontrar una lista de más de 500 colores. En la [ayuda de matplotlib.colors](https://matplotlib.org/stable/api/colors_api.html) puedes consultar toda la información sobre los colores de `plt.plot()`.
* **`marker`**. Los tipos de marcador se especifican igual que en el formato corto: `'s'`, `'o'`, `'^'`, `'*'`. Este argumento no tiene abreviatura. En [ayuda de matplotlib.markers](https://matplotlib.org/stable/api/markers_api.html)  puedes consultar los símbolos de los marcadores disponibles.
* **`linestyle/ls`**. Para el tipo de línea también se usan los mismos caracteres que en formato corto: `'-'`, `'--'`, `':'` y `'-.'`. Si no queremos que haya líneas, el valor de este argumento estará vacío (`linestyle=''`). En la [ayuda de Linestyles](https://matplotlib.org/stable/gallery/lines_bars_and_markers/linestyles.html) puedes encontrar más información sobre los tipos de línea.


Veamos algunos ejemplos con los argumentos en formato largo.

````python
plt.plot(x, y, color='black', marker='o', linestyle=':')
# equivalente a plt.plot(x, y, 'ko:') visto antes
plt.show()
````

````python
plt.plot(x, y, c='#c41e3a', marker='d', ls='-.')
# equivalente a plt.plot(x, y, 'rd-.')
plt.show()
````

### <font color='steelblue'>Otras especificaciones

Además de las características básicas como el color, el tipo de marcador y tipo de línea, existen otras que podemos configurar a través de argumentos de la función `plot()`. La mayoría de estos argumentos tiene también una abreviatura que podemos utilizar indistintamente, y que indicamos a continuación:

* **`alpha`** para controlar la opacidad (transparencia) del gráfico, con un valor entre 0 (transparencia total y no se ve nada) y 1 (opacidad total). Este argumento no tiene abreviatura.
* **`linewidth/lw`** para el grosor de la línea, indicado por un número mayor que cero (el cero no pinta línea).
* **`markersize/ms`** controla el tamaño de los marcadores, siendo 6 el tamaño por defecto. Se especifica con números positivos.
* **`markerfacecolor/mfc`** permite especificar el color de relleno de los marcadores, utilizando, igual que en `color`, el nombre del color en inglés o el código hexadecimal.
* **`markeredgewidth/mew`** controla el ancho del borde del marcador con un número. Si no se indica nada, su tamaño es 1. Si iguala o supera el tamaño del marcador, lo tapa.
* **`markeredgecolor/mec`** indica el color del borde del marcador,con las mismas opciones que `color`.


Veamos estos argumentos cómo funcionan en más ejemplos:

````python
# color, marcador, ancho de línea y opacidad
plt.plot(x, y, color='black', marker='s', linewidth=2)
plt.show()
````

````python
# color, marcador, ancho de línea y opacidad
plt.plot(x, y, color='black', marker='s', linewidth=2,alpha=0.5)
# equivalente a plt.plot(x, y, c='black', marker='s', lw=2,alpha=0.5)
plt.show()
````

````python
# tamaño del marcador
plt.plot(x, y, color='black', marker='s', linewidth=2, markersize=9)
# equivalente a plt.plot(x, y, c='black', marker='s', lw=2, ms=9)
plt.show()
````

````python
# Color de relleno del marcador
plt.plot(x, y, color='black', marker='s', linewidth=2, markersize=9,
         markerfacecolor='yellow')
# equivalente a plt.plot(x, y, c='black', marker='s', lw=2, ms=9,mfc='yellow')
plt.show()
````

````python
# Ancho del borde del marcador
plt.plot(x, y, color='black', marker='s', linewidth=2, markersize=9,
         markerfacecolor='yellow', markeredgewidth=3)
# equivalente a plt.plot(x, y, c='black', marker='s', lw=2, ms=9,mfc='yellow', mew=3)
plt.show()
````

````python
# Color del borde del marcador
plt.plot(x, y, color='black', marker='s', linewidth=2, markersize=9,
         markerfacecolor='yellow', markeredgewidth=3, markeredgecolor='blue')
# equivalente a plt.plot(x, y, c='black', marker='s', lw=2, ms=9,mfc='yellow', mew=3, mec='blue')
plt.show()
````

## <font color='steelblue'> 3. Varias representaciones en un gráfico

A veces queremos dibujar dos funciones diferentes en un mismo gráfico, generalmente para compararlas. Para ello basta con escribir dos veces la función `plot`, una con cada una de las funciones a representar, y solicitar después la visualización con la función `show`.

En nuestro ejemplo de representación de parábolas, vamos a representar el rango que va entre -5 y 5 para la variable `X`. Definimos pues `x` como una secuencia en este rango. Creamos a continuación las variables `y1` e `y2`, que recogen respectivamente los valores de la función $y=x^2 - 2$ y de $y=-x^2 + 2$ para los valores `x` fijos.

````python
# valores de "x"
x = list(range(-5,6))

# creamos las "y"s vacías
y1 = []
y2 = []
# Bucle recorriendo x
for i in x:
  # añadimos los valores
  y1.append(i**2 - 2)
  y2.append(-i**2 + 2)

# vemos los valores
print(x)
print(y1)
print(y2)
````

Una vez hemos calculado los valores de las funciones en varios puntos, ya podemos representarlos en el mismo gráfico:

````python
import matplotlib.pyplot as plt
%matplotlib inline

# Función 1
plt.plot(x, y1)
# Función 2
plt.plot(x, y2)

# Mostramos el gráfico
plt.show()
````

Si no indicamos nada en los argumentos, la primera función se representa en color azul y la segunda en color naranja.

Vamos a personalizar ambas funciones, para diferenciarlas como deseemos.

````python
# Función 1 en rojo, marcadores rojos y línea continua
plt.plot(x, y1, 'ro-')
# Función 2 en negro, marcadores cuadrados y línea punteada
plt.plot(x, y2, 'ks:')

# Mostramos el gráfico
plt.show()
````

Por supuesto, podemos combinar el formato corto o largo en la especificación de los argumentos de `plot()`.

````python
# Función 1
plt.plot(x, y1, 'ro-')
# Función 2
plt.plot(x, y2, color='black', marker='s', linestyle=':', linewidth=0.5)

# Mostramos el gráfico
plt.show()
````

### <font color='steelblue'>Leyendas

Cuando representamos datos o funciones, es importante identificar qué representa cada color, forma o tamaño, y para ello utilizamos las leyendas. Las leyendas se generan con la **función `plt.legend()`** del módulo `matplotlib.pyplot`, antes de mostrar la figura (con `plt.show()`) y después de dibujarlas (con `plt.plot()`).

Si utilizamos simplemente la función `plt.legend()`sin especificar qué etiquetas queremos mostrar en la leyenda, obtendremos un *warning* avisándonos de ello. Veámoslo con el ejemplo anterior.

````python
# Función 1
plt.plot(x, y1, 'ro-')
# Función 2
plt.plot(x, y2, color='black', marker='s', linestyle=':', linewidth=0.5)
# Indicamos la leyenda
plt.legend()
# Mostramos el gráfico
plt.show()
````

Hemos pues de especificar las etiquetas que deseamos que aparezcan en la leyenda a través del **argumento `label`** en la **función `plt.plot()`**. Cada función dibujada con `plt.plot()` mostrará en la leyenda la cadena de caracteres que hayamos definido en `label`.  

Con el ejemplo anterior, vamos a identificar las dos funciones representadas con las etiquetas 'y1' e 'y2'. Veámoslo.

````python
# Función 1
plt.plot(x, y1, 'ro-', label='y1')
# Función 2
plt.plot(x, y2, color='black', marker='s', linestyle=':', linewidth=0.5, label='y2')
# Indicamos la leyenda
plt.legend()
# Mostramos el gráfico
plt.show()
````

Cuando queremos que las etiquetas contengan símbolos matemáticos (por ejemplo, la definición de las funciones que representan), hemos de utilizar notación $\LaTeX$, escribiendo entre comillas la fórmula matemática encajada entre dos símbolos de dólar, `$...$`. Lo hacemos a continuación.

````python
# Función 1
plt.plot(x, y1, 'ro-', label='$x^2 - 2$')
# Función 2
plt.plot(x, y2, color='black', marker='s', linestyle=':', linewidth=0.5, label='$-x^2 + 2$')
# Indicamos la leyenda
plt.legend()
# Mostramos el gráfico
plt.show()
````

### <font color='steelblue'>Posición de la leyenda

Por defecto, la leyenda se suele ubicar, automáticamente, en algún lugar del gráfico en el que no interfiere con las líneas y puntos representados. Sin embargo, podemos indicar dónde queremos que aparezca. Para ello utilizaremos el **argumento `loc=`** dentro de la **función `plt.legend()`**. Los valores que acepta este argumento son: `'upper right'`, `'lower right'`, `'upper left'`, `'lower left'`, `'upper center'`, `'lower center'`, `'center left'`, `'center right'`, `'center'` y `'best'`. La opción `'best'` corresponde con la ubicación por defecto de `matplotlib`.

En la [ayuda de `legend`](https://matplotlib.org/stable/api/legend_api.html) puedes encontrar más información sobre las leyendas.

Probemos algunas opciones de posicionamiento.

````python
# Función 1
plt.plot(x, y1, 'ro-', label='$x^2 - 2$')
# Función 2
plt.plot(x, y2, color='black', marker='s', linestyle=':', linewidth=0.5, label='$-x^2 + 2$')
# Colocamos la leyenda arriba a la derecha
plt.legend(loc='upper right')
# Mostramos el gráfico
plt.show()
````

````python
# Función 1
plt.plot(x, y1, 'ro-', label='$x^2 - 2$')
# Función 2
plt.plot(x, y2, color='black', marker='s', linestyle=':', linewidth=0.5, label='$-x^2 + 2$')
# Colocamos la leyenda abajo a la izquierda
plt.legend(loc='lower left')
# Mostramos el gráfico
plt.show()
````

````python
# Función 1
plt.plot(x, y1, 'ro-', label='$x^2 - 2$')
# Función 2
plt.plot(x, y2, color='black', marker='s', linestyle=':', linewidth=0.5, label='$-x^2 + 2$')
# Colocamos la leyenda arriba en el centro
plt.legend(loc='upper center')
# Mostramos el gráfico
plt.show()
````

## <font color='steelblue'> 4. Varios gráficos en un lienzo

En ocasiones, en lugar de superponer dos representaciones para compararlas, preferimos presentarlas en gráficos/ventanas distintas, pero visualizadas a la vez en el mismo lienzo. Para ello hemos de utilizar la **función `plt.subplots()`** del módulo `matplotlib`.


Esta función, que se utiliza para cada uno de los gráficos que queremos visualizar en un mismo lienzo, tiene tres argumentos,
```plt.subplots(filas,columnas,posición)```
- el primero indica el número de **filas** del lienzo en el que ubicaremos los diferentes gráficos,
- el segundo argumento indica el número de **columnas**
- el tercero indica la **posición** que pretendemos para el gráfico a representar a continuación con `plt.plot()`. Se trata de un número entero que marca el orden en la secuencia de gráficos que se suceden, recorriendo filas y columnas de izquierda a derecha y de arriba a abajo.

A continuación de la función `plt.subplot()`, dibujamos el gráfico correspondiente con `plt.plot()`. Repetimos el proceso hasta ubicar todos los gráficos que queremos presentar en el mismo lienzo.

Veamos ahora cómo lo hacemos.

````python
# Una fila y dos columnas: subplot(1,2, )

# primer subgráfico
plt.subplot(1,2,1)
plt.plot(x, y1, 'ro-')

# segundo subgráfico
plt.subplot(1,2,2)
plt.plot(x, y2, 'ks:')

plt.show()
````

````python
# Dos filas y una columna: subplot(2,1, )

# primer subgráfico
plt.subplot(2,1,1)
plt.plot(x, y1, 'ro-')

# segundo subgráfico
plt.subplot(2,1,2)
plt.plot(x, y2, 'ks:')

plt.show()
````

También se pueden crear figuras con más subgráficos, es decir, con más filas y más columnas.

Como ejemplo, incorporamos dos funciones más: $y3=x^2 + 5$ e $y4=-x^2 - 5$, para lo que hemos de crear primero las nuevas listas con los valores de los resultados de las funciones.

````python
# creamos las "y"s vacías
y3 = []
y4 = []
for i in x:
  # añadimos los valores
  y3.append(i**2 + 5)
  y4.append(-i**2 - 5)

# verificamos las listas
print(x)
print(y3)
print(y4)
````

Ya podemos representar:

````python
# Dos filas y dos columnas: subplot(2,2,)

# primer subgráfico
plt.subplot(2,2,1)
plt.plot(x, y1, 'ro-')

# segundo subgráfico
plt.subplot(2,2,2)
plt.plot(x, y2, 'ks:')

# tercer subgráfico
plt.subplot(2,2,3)
plt.plot(x, y3, 'g^-.')

# cuarto subgráfico
plt.subplot(2,2,4)
plt.plot(x, y4, 'b*--')

plt.show()
````

````python
# Una fila y cuatro columnas: subplot(1,4, )

# primer subgráfico
plt.subplot(1,4,1)
plt.plot(x, y1, 'ro-')

# segundo subgráfico
plt.subplot(1,4,2)
plt.plot(x, y2, 'ks:')

# tercer subgráfico
plt.subplot(1,4,3)
plt.plot(x, y3, 'g^-.')

# cuarto subgráfico
plt.subplot(1,4,4)
plt.plot(x, y4, 'b*--')

plt.show()
````

La distribución de los gráficos no tiene por qué ser simétrica y podemos dejar huecos:

````python
# Dos filas y tres columnas: subplot(2,3, )

# primer subgráfico
plt.subplot(2,3,1)
plt.plot(x, y1, 'ro-')

# segundo subgráfico
plt.subplot(2,3,2)
plt.plot(x, y2, 'ks:')

# tercer subgráfico
plt.subplot(2,3,4)
plt.plot(x, y3, 'g^-.')

# cuarto subgráfico
plt.subplot(2,3,6)
plt.plot(x, y4, 'b*--')

plt.show()
````

````python
# Dos filas y tres columnas: subplot(2,3, )

# primer subgráfico
plt.subplot(2,3,1)
plt.plot(x, y1, 'ro-')

# segundo subgráfico
plt.subplot(2,3,2)
plt.plot(x, y2, 'ks:')

# tercer subgráfico
plt.subplot(2,3,5)
plt.plot(x, y3, 'g^-.')

# cuarto subgráfico
plt.subplot(2,3,6)
plt.plot(x, y4, 'b*--')

plt.show()
````

### <font color='steelblue'>Gráficos dentro de gráficos

En ocasiones nos interesa anidar (encajar) un gráfico dentro de otro.
Para ello utilizaremos la **función `plt.axes()`**, que utilizaremos para posicionar un gráfico antes de pintarlo con `plt.plot()`. Los argumentos que recibe en cada ejecución son las coordenadas del marco en el que deseamos encuadrar cada gráfico, variando todas ellas entre 0 y 1 (recomendando evitar los extremos):

```plt.axes([xmin,ymin,xmax,ymax])```

* **xmin**: la posición *x* de la esquina inferior izquierda.
* **ymin**: la posición *y* de la esquina inferior izquierda.
* **xmax**: la posición *x*, respecto de xmin, de la esquina superior derecha.
* **ymax**: la posición *y*, respecto de ymin, de la esquina superior derecha.

Con estas cuatro coordenadas (dos puntos), se dibuja el contorno del gráfico.

Para integrar el gráfico interior, hemos de asegurarnos que sus coordenadas son menores a las del gráfico contenedor. Este gráfico que se cree, se superpondrá al que ya existe.

Tras cada ejecución de `axes()`, ejecutamos el `plot()` correspondiente, y por último, como siempre, mostraremos el gráfico con `show()`.

Veamos un ejemplo con *y1* e *y2*:

````python
# gráfico contenedor
plt.axes([0.05,0.05,0.95,0.95])
plt.plot(x, y2, 'ro-')

# gráfico interior
plt.axes([0.3,0.3,0.35,0.35])
plt.plot(x, y1, 'ks:')

plt.show()
````

## <font color='steelblue'> 5. Configuración del entorno gráfico

Vemos a continuación cómo configurar otros elementos y características que conforman el entorno gráfico que mostramos al representar los objetos matemáticos que nos interesan.

### <font color='steelblue'>Etiquetas para título y ejes

Empezaremos con las **funciones `plt.title()`, `plt.xlabel()` e `plt.ylabel()`**, con las que especificamos, respectivamente, el título del gráfico, el del eje de abscisas (eje *X*), y el del eje de ordenadas (eje *Y*). El argumento que reciben estas funciones es una cadena de texto, y por tanto irá entrecomillada.

Además de la etiqueta, podemos definir otras opciones adicionales, como la fuente, el tamaño de letra, ubicación, etc, que podemos consultar en la [ayuda de `title`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.title.html).

Empezamos dando un título a la función *y1*: "Primera función".

````python
plt.plot(x, y1, 'ro-')
# ponemos el título
plt.title('Primera función')

plt.show()
````

Utilizamos ahora `fontsize` para el tamaño de letra y `loc` para la colocación:

````python
plt.plot(x, y1, 'ro-')
# ponemos el título
plt.title('Primera función', fontsize=20, loc='left')

plt.show()
````

Añadimos etiquetas a los ejes $X$ e $Y$.

````python
plt.plot(x, y1, 'ro-')
# ponemos el título
plt.title('Primera función', fontsize=20)
# ponemos el eje x
plt.xlabel('Eje x', fontsize=10)
# ponemos el eje y
plt.ylabel('Eje y', fontsize=10)

plt.show()
````

Podemos incluir expresiones matemáticas en estas etiquetas, utilizando la sintaxis de $\LaTeX$ y ubicando las fórmulas entre dos símbolos `$`.

````python
plt.plot(x, y1, 'ro-')
# ponemos el título
plt.title('Función $y = x^2 - 2$', fontsize=20)
# ponemos el eje x
plt.xlabel('Eje x', fontsize=10)
# ponemos el eje y
plt.ylabel('Eje y', fontsize=10)

plt.show()
````

### <font color='steelblue'>Amplitud de los ejes

En ocasiones queremos enfocar un gráfico aplicando un zoom de aumento o de decremento, o lo que es lo mismo, visualizando el gráfico solo en un rango de valores más pequeño. Esto lo conseguimos modificando el rango de valores $x$ e $y$ que visualizamos, y lo hacemos con las **funciones `plt.xlim()` y `plt.ylim()`**, respectivamente para el eje *X* y el eje *Y*. Podemos indicar como argumentos el mínimo valor a visualizar los límites en X **(`xmin=`,`xmax=`)**, o en Y **(`ymin=`,`ymax=`)** o ambos, en cuyo caso no es preciso indicar a qué argumento nos referimos, pero sí utilizar el orden **`(min,max)`**.

Como ejemplo, queremos visualizar en el gráfico de `y1` los valores del 0 al 4 en el eje *X* y como valor máximo en el eje *Y* el 10.

````python
plt.plot(x, y1, 'ro-')
plt.title('Función $y = x^2 - 2$', fontsize=20)
plt.xlabel('Eje x', fontsize=10)
plt.ylabel('Eje y', fontsize=10)
# delimitador de rango en el eje x
plt.xlim(0,4)
# delimitador de rango en el eje y
plt.ylim(ymax=10)

plt.show()
````

**Nota:** Si escribimos los límites en orden inverso, (max,min), el gráfico se invierte (al ir de mayor a menor), como si viéramos su reflejo en un espejo.

Al dibujar con la configuración por defecto, `plt.plot()` consigue un encuadre perfecto de los ejes (equivalente más o menos, a multiplicar por 1.1 los valores mínimos y máximo del rango de valores definidos).

````python
plt.plot(x, y1, 'ro-')
plt.title('Función $y = x^2 - 2$', fontsize=20)
plt.xlabel('Eje x', fontsize=10)
plt.ylabel('Eje y', fontsize=10)
# delimitador de rango en el eje x
plt.xlim(min(x)*1.1, max(x)*1.1)
#delimitador de rango en el eje y
plt.ylim(min(y1)*1.1, max(y1)*1.1)

plt.show()
````

###  <font color='steelblue'> Marcas de los ejes

Podemos también indicar qué marcas y qué números/etiquetas queremos que aparezcan en cada uno de los ejes. Para ello utilizaremos las **funciones `plt.xticks()` y `plt.yticks()`**, que han de recibir como argumento principal una lista con los valores en los que queremos mostrar marcas.

Por ejemplo, queremos que solo aparezcan marcas en el eje $X$ en los valores $(-4,0,4)$ y en el eje *Y* en $(0,10,20)$.

````python
plt.plot(x, y1, 'ro-')
plt.title('Función $y = x^2 - 2$', fontsize=20)
plt.xlabel('Eje x', fontsize=10)
plt.ylabel('Eje y', fontsize=10)
plt.xlim(min(x)*1.1, max(x)*1.1)
plt.ylim(min(y1)*1.1, max(y1)*1.1)
# eje x
plt.xticks([-4,0,4])
# eje y
plt.yticks([0,10,20])

plt.show()
````

También podemos definir nuestras propias etiquetas para cada una de las marcas, incluyendo como segundo argumento una lista con las etiquetas que deseamos, bien sean textos o expresiones matemáticas entrecomilladas.

Tocamos las etiquetas de las marcas a continuación.

````python
plt.plot(x, y1, 'ro-')
plt.title('Función $y = x^2 - 2$', fontsize=20)
plt.xlabel('Eje x', fontsize=10)
plt.ylabel('Eje y', fontsize=10)
plt.xlim(min(x)*1.1, max(x)*1.1)
plt.ylim(min(y1)*1.1, max(y1)*1.1)
# eje x
plt.xticks([-4,0,4], ['-4m', '0m', '4m'])
# eje y
plt.yticks([0,10,20], ['$0m^2$', '$10m^2$', '$20m^2$'])

plt.show()
````

### <font color='steelblue'>Espacios entre gráficos

Cuando añadimos etiquetas a los ejes y títulos a los gráficos y los mostramos agrupados en un mismo lienzo, nos puede pasar algo como lo que vemos en la siguiente figura: que se superponen títulos y etiquetas y no se pueden leer.

````python
# Amontonamiento (por defecto) de gráficos en un mismo lienzo
plt.subplot(2,2,1)
plt.plot(x, y1, 'ro-')
plt.title('Primera función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.subplot(2,2,2)
plt.plot(x, y2, 'ks:')
plt.title('Segunda función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.subplot(2,2,3)
plt.plot(x, y3, 'g^-.')
plt.title('Tercera función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.subplot(2,2,4)
plt.plot(x, y4, 'b*-')
plt.title('Cuarta función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.show()
````

Para evitar este inconveniente tenemos como opción ampliar automáticamente el espacio entre los gráficos con la **función `plt.tight_layout()`**. Esta función ajusta automáticamente los espacios entre gráficos para no solapar etiquetas, se utiliza después de crear todos los gráficos a ubicar en el mismo lienzo, y siempre antes de mostrarlo con `plt.show()`:

````python
# Separación automática de los gráficos en un lienzo
plt.subplot(2,2,1)
plt.plot(x, y1, 'ro-')
plt.title('Primera función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.subplot(2,2,2)
plt.plot(x, y2, 'ks:')
plt.title('Segunda función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.subplot(2,2,3)
plt.plot(x, y3, 'g^-.')
plt.title('Tercera función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.subplot(2,2,4)
plt.plot(x, y4, 'b*-')
plt.title('Cuarta función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

# ensanchamos
plt.tight_layout()

plt.show()
````

### <font color='steelblue'>Tamaño de un lienzo

Hasta ahora hemos comprobado cómo la función `plt.show()` genera una figura o lienzo siempre con el mismo tamaño, independientemente de los gráficos que contenga, y adaptando el tamaño de estos al lienzo. Si queremos modificar el tamaño por defecto del lienzo o figura final, podemos hacerlo con la **función `plt.figure()`**, que recibe dos argumentos:

* **`figsize`**: una tupla que indica el tamaño (ancho y alto) del lienzo en pulgadas (por defecto utiliza 6.4x4.8).
* **`dpi`**: un entero que indica el número de puntos por pulgada (por defecto utiliza 100), y hace referencia al tamaño del contenido en pantalla.

Esta función se ejecuta antes de crear todos los gráficos.

En https://matplotlib.org/stable/api/figure_api.html puedes encontrar más información sobre la funcion figure().



A continuación varios ejemplos en los que modificamos ambos argumentos.

````python
# Cambiamos el tamaño del lienzo a 10x7 dejando los mismos puntos por pulgada
plt.figure(figsize=(10,7))

plt.subplot(2,2,1)
plt.plot(x, y1, 'ro-')
plt.title('Primera función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.subplot(2,2,2)
plt.plot(x, y2, 'ks:')
plt.title('Segunda función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.subplot(2,2,3)
plt.plot(x, y3, 'g^-.')
plt.title('Tercera función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.subplot(2,2,4)
plt.plot(x, y4, 'b*--')
plt.title('Cuarta función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.tight_layout()

plt.show()
````

````python
# Cambiamos el tamaño del lienzo a 10x7 y reducimos a 20 los puntos por pulgada
plt.figure(figsize=(10,7), dpi=20)

plt.subplot(2,2,1)
plt.plot(x, y1, 'ro-')
plt.title('Primera función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.subplot(2,2,2)
plt.plot(x, y2, 'ks:')
plt.title('Segunda función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.subplot(2,2,3)
plt.plot(x, y3, 'g^-.')
plt.title('Tercera función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.subplot(2,2,4)
plt.plot(x, y4, 'b*--')
plt.title('Cuarta función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.tight_layout()

plt.show()
````

### <font color='steelblue'>Almacenamiento de figuras

El módulo `matplotlib` permite guardar las figuras creadas en ficheros de alta calidad y en varios formatos, incluyendo PNG, JPG, EPS, SVG, PGF y PDF.

Para guardar una figura en un archivo podemos disponemos de la **función `plt.savefig()`**, que recibe como argumento la cadena de texto con el nombre que queremos dar al archivo (argumento **`fname`**), incluida la extensión para el formato deseado.

También podemos especificar la resolución con el argumento **`dpi`** y modificar otras opciones sobre las que puedes ampliar información en la [ayuda de `savefig`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.savefig.html).

Esta función `plt.savefig()` se ejecuta después la creación de todos los gráficos del lienzo y antes de mostrarlo por pantalla con `plt.show()`.

Si no estamos conectados a *Google Drive*, se guardará en el directorio */content/* de nuestra máquina virtual. Si está montada la unidad de Drive, basta con especificar la ruta deseada.

Para ejemplificar su uso utilizamos de nuevo la figura que hemos generado anteriormente y la guardamos en formato 'png' con el nombre 'grafico1.png'.

````python
# Almacenaje de un gráfico png
plt.figure(figsize=(10,7), dpi=80)

plt.subplot(2,2,1)
plt.plot(x, y1, 'ro-')
plt.title('Primera función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.subplot(2,2,2)
plt.plot(x, y2, 'ks:')
plt.title('Segunda función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.subplot(2,2,3)
plt.plot(x, y3, 'g^-.')
plt.title('Tercera función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.subplot(2,2,4)
plt.plot(x, y4, 'b*--')
plt.title('Cuarta función')
plt.xlabel('Eje x')
plt.ylabel('Eje y')

plt.tight_layout()

# guardamos la figura
plt.savefig("grafico1.png")

plt.show()
````
