### <font color='steelblue'>Estructuras de datos en Pandas

El módulo de **Python** llamado _Pandas_ proporciona unas estructuras de datos rápidas, flexibles y expresivas, diseñadas para que el trabajo con datos _relacionales_ o _etiquetados_ sea fácil e intuitivo.

_Pandas_ es adecuado para el tratamiento de diferentes tipos de datos:

- Datos tabulares con columnas de tipo heterogéneo, como en una tabla de SQL o una hoja de cálculo de Excel.
- Datos de series temporales, ordenados y no ordenados (no necesariamente de frecuencia fija).
- Matriz de datos arbitraria, con elementos homogéneos o  heterogéneos, y con etiquetas de filas y/o columnas.
- Cualquier otra forma de conjuntos de datos observacionales / estadísticos.

Este módulo nos ofrece las siguientes estructuras de datos:

* __Series__: se trata de _arrays_ unidimensionales indexados (con índice o etiqueta), similares a los diccionarios. Se pueden generar a partir de diccionarios o de listas.

* __DataFrame__: son estructuras de datos similares a las tablas de bases de datos relacionales (como SQL, o más sencillo Excel). También se pueden crear a partir de listas o diccionarios.

* __Panel, Panel4D y PanelND__: Estas estructuras de datos permiten trabajar con más de dos dimensiones. Dado que son estructuras complejas no trataremos los paneles en estos cuadernos.

Cuando tenemos un objeto definido con *Pandas*, para conocer de qué tipo (de los anteriores) es, utilizamos el comando:
```
**type(objeto)**
```

Algunas de las **ventajas** de usar _Pandas_ son:

- Facilita el manejo de grandes bases de datos, con datos de diferentes tipos, y en particular de los _datos faltantes_.

- Permite modificar de modo sencillo el tamaño de los objetos, como _insertar y eliminar_ columnas en un _DataFrame_, o realizar recortes y extraer subconjuntos de datos.

- Permite asignar de modo sencillo etiquetas a conjuntos de datos, automática o explícitamente.

- Proporciona herramientas robustas para cargar distintos tipos de archivos de datos (CSV, EXCEL, bases de datos, e incluso _HDF5 ultrarrápido_).

- Dispone de una gran cantidad de funciones para manipular datos agrupadas por funcionalidad (limpiar, transformar, visualizar,...).

- Sus funciones son muy eficientes para la manipulación de datos, y se integran fácilmente con otras librerías de Python.

- Tiene múltiples funcionalidades para tratamiento de datos de series temporales (fechas, frecuencias, ventanas temporales, ...).

- Permite transformar datos no estructurados para convertirlos en datos procesables. Los datos no estructurados son aquellos que no tienen una estructura interna identificable y por tanto carecen de valor hasta que se organizan (como por ejemplo las publicaciones en redes sociales).

Como desventajas cabe mencionar que tiene una curva de aprendizaje pronunciada al inicio, y que puede requerir de mucha memoria en el computador cuando se manipulan grandes bases de datos.

En conclusión, el módulo _Pandas_ es la herramienta base y más popular para realizar todas las tareas asumidas por los científicos de datos, desde el muestreo y la limpieza de datos, pasando por el análisis o modelado, hasta la visualización y la presentación de resultados.

### <font color='steelblue'>Tipos de variables</font>

En el ámbito estadístico hablamos de **variables** cuando tenemos que referirnos a características que observamos en la población de interés con algún fin informativo, explicativo o incluso predictivo. Así, para describir o explicar el gasto de las familias en nuestro país, podríamos recopilar el número de coches que tiene cada familia, los ingresos mensuales, cuánto se gasta cada semana a lo largo del año en la cesta de la compra, e incluso registrar qué productos componen dicha cesta. Todas las características que registramos para un estudio son variables y, como en el ejemplo, pueden ser de distintos tipos, hecho que influye en cómo las almacenamos y operamos con ellas.

Así pues, podemos hablar de distintos **tipos de datos**, o de **variables** que _Python_ identifica de modos distintos, y que se muestran en la tabla a continuación para los ejemplos que hemos planteado anteriormente:

| Tipo de dato/variable | Identificador  | Ejemplo |
| :-------------------- | :------------- | :------------- |
| **Cadena de texto** | `object (O)` | productos de la cesta de la compra |
| **Número entero** | `int64`    | número de coches en la familia |
| **Número decimal** | `float64` | ingresos mensuales |
| **Series temporales** | `datetime64` o `timedelta` | gasto semanal en la cesta de la compra |

El número '64' en los identificadores de enteros, decimales y series temporales, indica el número de bits de memoria que se asignan para almacenar datos en una celda.

Cuando leemos una base de datos recopilados sobre una muestra, a veces las variables registradas no tienen bien identificado el tipo de dato que almacena, por lo que siempre será preciso una revisión exhaustiva. Para ello existen varios métodos:

* **`.dtype`**, que aplicado a una variable, devuelve el identificador del tipo de dato que contiene.

Cuando agrupamos varias variables, por columnas, en un _dataframe_, podemos preguntar globalmente por el tipo de datos en todas las columnas con:

* **`.dtypes`**, que aplicado a un dataframe devuelve el tipo de dato para cada una de las variables/columnas.




En Estadística además, siempre hacemos otra clasificación (alternativa) de las variables, con la que tratamos de detallar, básicamente, cómo es la información que hemos recopilado, esto es, si se refiere a números, o a cualidades. Hablamos entonces de una diferenciación de las variables según dos grandes grupos, que a su vez se diferencia en otros subgrupos:

* **Variables cuantitativas:** cuyas observaciones registramos con números que tienen sentido como números, y que a su vez pueden ser:
  * __cuantitativas discretas__, que se expresan con números enteros (como el número de hijos, el número de trasplantes, el número de colegios en la comarca, ...)
  * __cuantitativas continuas__: números que aceptan decimales (como el bmi, el peso, la distancia entre hospitales, ...).

* **Variables cualitativas**, que nos propocionan información sobre una cualidad, una percepción, etc., esto es, no se responden/observan con un número, y nos permiten clasificar a los sujetos que estudiamos en grupos. Son variables de este tipo el color de ojos, la localidad de residencia, las respuestas a una encuesta de satisfacción, etc. Estos datos pueden ser cadenas de texto, pero también son representables mediante números enteros, y a veces con series temporales. Se distinguen dos subgrupos:

  * __cualitativas ordinales__, en las que las respuestas o valores observables posibles tienen un orden intrínseco, como en las encuestas de satisfacción las respuestas desde Muy insatisfecho a Muy satisfecho, y que a veces se plantean con una pregunta del tipo *'Puntúa del 1 al 5'*  para obtener ya directamente la respuesta codificada y ahorrar trabajo de codificación.
  
  * __cualitativas nominales__, en las que la respuesta es necesariamente una cadena de texto, como el color de ojos, o la localidad de residencia.

Las variables cualitativas son básicamente las variables de cadena de texto en Python, si bien para trabajar con ellas es habitual **codificar** las respuestas posibles con números que se utilizan como etiquetas para simplificar las respuestas posibles y facilitar el análisis. Así, el color de ojos se puede codificar, por ejemplo, con 1 para el verde, 2 para el azul, 3 para el marrón, y 4 para el resto, y esta codificación es arbitraria, generalmente decidida por el analista.

A partir de estas definiciones, encontraremos casos en los que variables cuantitativas discretas las tratemos como cuantitativas continuas (por ejemplo cuando calculamos el número medio de hijos en España), o incluso como cualitativas nominales (si nos interesa el porcentaje de sujetos que no tienen hijos frente a los que tienen al menos uno).
  
Una mención especial merecen las __variables cualitativas binarias o dicotómicas__, que son aquellas en las que exclusivamente se registran dos respuestas (cualitativas) posibles, como por ejemplo si un paciente es diabético o no. En estos tipos de variables se suele optar por una codificación 0/1.

Generalmente, cuando accedemos a una base de datos recopilada y almacenada por otros, se suele proporcionar una breve explicación del significado, respuestas posibles y la codificación de cada una de las variables observadas. Esta información nos permitirá asegurarnos de si efectivamente los tipos de datos que reconoce Python han sido bien registrados, o si no ha sido así, abordar las modificaciones pertinentes antes de iniciar el trabajo, para no errar en el análisis.

## <font color='steelblue'> 2. Dataframes

Un _dataframe_ de la librería _Pandas_ es el tipo de objeto útil para representar una base de datos con una estructura de matriz bidimensional, en la que las columnas representan variables y en cada fila se distribuyen los registros de todas las variables en cada uno de los sujetos u objetos observados o medidos. Estas filas, como en las series, están indexadas y los índices pueden ser sustituidos por etiquetas. Básicamente viene a representar una agrupación de series por columnas, con lo cual puede contener columnas de distintos tipos (texto, enteros, decimales,...), pero cada columna será de un único tipo. Por supuesto, todas las series agrupadas en un _dataframe_, esto es, todas las columnas, han de tener la misma longitud o número de elementos.

### <font color='steelblue'>Crear  </font>



Podemos crear _dataframes_ a partir de listas, arrays, y también a partir de diccionarios, con la función **`DataFrame()`** de *Pandas*. Además, podemos cargar como _dataframes_ archivos en diferentes formatos, disponibles en nuestro espacio de trabajo o en internet.



####<font color='steelblue'> Dataframes con listas

Para **crear _dataframes_ a partir de listas** hemos de:
1. agrupar las listas que representan las columnas en un array bidimensional, con **`np.stack()`**
1. definir dicha agrupación como dataframe con  **`pd.DataFrame()`**.

Vamos a generar una base de datos con la temperatura media diaria, durante una semana, de las capitales de provincia Alicante, Castellón y Valencia. Seguimos los pasos anteriores:

````python
# Definimos como listas los valores/registros de temperatura de cada ciudad
Alicante = [23.3, 19.8, 16.3, 14, 13.5, 13.4, 12.6]
Castellon = [20.9, 15.7, 15.3, 12.6, 13.3, 14.3, 14.1]
Valencia = [21.3, 15.3, 13.7, 10.7, 10.8, 8.5, 11.9]

#agrupamos por columnas las listas definidas
temperaturas = np.stack([Alicante, Castellon, Valencia], axis=1)

# definimos dicha agrupación como DataFrame
temperatura = pd.DataFrame(temperaturas)
print(temperatura)
````

Comprobamos que la primera columna contiene los datos registrados para Alicante, la segunda de Castellón y la tercera de Valencia.

Si no indicamos nombres para las columnas, estas quedan  nombradas con números enteros correlativos que empiezan en 0. Podemos sin embargo, dar un nombre a cada columna asignando una lista (de nombres) al argumento `columns` de la función `DataFrame()`.

Repetimos la definición del dataframe, asignando nombres a las columnas a través del argumento **`columns^** en la función `DataFrame()`.

````python
# Creamos el dataframe de nuevo, indicando un nombre para las columnas
temperatura = pd.DataFrame(temperaturas, columns=["Alicante", "Castellón", "Valencia"])
print(temperatura)
````

Por defecto, las filas quedan indexadas numéricamente con la secuencia de enteros que inician en 0:
<center><img src=https://raw.githubusercontent.com/ia4legos/Statistics/main/images/img_160_30_10.png width="350" height="325"></center>



Al igual que podemos incluir los nombres de las columnas al crear el dataframe, podemos definir los nombres para los índices con el argumento **`index`** en `pd.DataFrame()`.

Definimos a continuación una lista con las etiquetas para los índices, y los incorporamos al crear el _dataframe_.


````python
indices = ["lunes","martes","miércoles","jueves","viernes","sábado","domingo"]

# definimos dicha agrupación como DataFrame
temperatura = pd.DataFrame(temperaturas,
                          columns=["Alicante", "Castellón", "Valencia"],
                          index = indices)
print(temperatura)
````

####<font color='steelblue'> Dataframes con diccionarios

También podemos **crear un _dataframe_ a partir de un diccionario**, resultando incluso más sencillo que con listas. Para ello basta con:
1. definir un diccionario donde la clave (_key_) es el nombre de la variable/columna en el _dataframe_, y los valores registrados vienen dados en una lista;
2. transformar el diccionario en un dataframe con la función `pd.DataFrame()`.

Como ejemplo vamos a reproducir el _dataframe_ anterior a través de un diccionario.

````python
# definimos un diccionario
temperaturas = {"Alicante": [23.3, 19.8, 16.3, 14, 13.5, 13.4, 12.6],
                "Castellon": [20.9, 15.7, 15.3, 12.6, 13.3, 14.3, 14.1],
                "Valencia": [21.3, 15.3, 13.7, 10.7, 10.8, 8.5, 11.9]}
indices = ["lunes","martes","miércoles","jueves","viernes","sábado","domingo"]

# transformamos el diccionario a dataframe
temperatura = pd.DataFrame(temperaturas,index = indices)
print(temperatura)
````

#### <font color='steelblue'>Dataframes desde archivos

Muchas veces las bases de datos con las que trabajamos proceden de otras fuentes y vienen dadas en ficheros de distintos tipos, con extensiones _csv_, _xlsx_, _html_, etc. Con _Pandas_ podemos leer fácilmente estos archivos y convertirlos directamente en _dataframes_ para poderlos manipular de forma eficiente. * **`read_json`**: para ficheros en JavaScript con extensión _json_;
* **`read_html`**: lee tablas en formato _html_ y las transforma en listas o _dataframes_;
* **`read_sql`**: lee consultas o bases de datos de _SQL_.

Para ampliar información sobre el funcionamiento de estas funciones y consultar cómo leer otros tipos de archivos, te recomendamos consultar el [tutorial de Pandas](https://pandas.pydata.org/pandas-docs/stable/user_guide/io.html).

El segundo archivo, 'diabetes.csv', contiene información clínica de 768 pacientes respecto a 9 variables diferentes:

* _`pregnancies`_: variable cuantitativa discreta; indica el número de embarazos de la paciente
* _`glucose`_: variable cuantitativa discreta; proporciona los miligramos (_ml_ sin decimales) de glucosa en sangre cuando acudió a la consulta;

* _`insulin`_: variable cuantitativa discreta; son los miligramos de insulina en sangre (_mg_ sin decimales);
* _`bmi`_: variable cuantitativa continua; muestra el índice de masa corporal (con decimales);
* _`dpf`_: variable cuantitativa continua; _diabetes pedigree function_
* _`age`_: variable cuantitativa discreta; da la edad de la paciente;
* _`diabetes`_: variable cualitativa binaria o dicotómica, con 1 si tiene diabetes y 0 si no la tiene.



````python
# Lectura del fichero y creación de un dataframe
diabetes = pd.read_csv("diabetes.csv")
# comprobamos que la lectura es correcta
print(diabetes)
# comprobamos que es un dataframe
print(type(diabetes))
# comprobamos el tipo de las variables en el dataframe
print(diabetes.dtypes)
````

No obstante, a veces los separadores son otros, y entonces el ajuste por defecto no proporciona una lectura óptima para poder trabajar con la base de datos. Prueba ahora a abrir y visualizar desde el directorio de trabajo el fichero `encuesta_nutricion_cv.csv`, y también prueba a leer como antes el fichero.

````python
# Lectura del fichero con los ajustes por defecto
encuesta = pd.read_csv("encuesta_nutricion_cv.csv")
# comprobamos qué se ha leído
print(encuesta)
````

Al leer este archivo como los anteriores, comprobamos que todos los datos en una fila aparecen apiñados y separados por el punto y coma (`;`), de modo que no se diferencian variables ni valores: el dataframe no es operativo así. Para que la lectura sea correcta, debemos utilizar como argumento de `read_csv` el argumento **`sep=";"`**, que separa los valores de las diferentes variables en cada fila. Observa también que la primera línea contiene los nombres de las variables, separados por `;`.

Es importante pues, antes de leer un fichero, intentar visualizarlo para comprobar en qué formato viene, cómo se separan los valores y cuál es el carácter utilizado para separar las cifras decimales. Todas estas cuestiones las podemos ajustar como argumentos al llamar a la función `read_csv`.

Cuando trabajamos sobre una base de datos será frecuente la selección de elementos, filas, columnas o una combinación de ambas, para ir resolviendo cuestiones de interés. Aprendemos aquí los modos básicos de selección en dataframes, tanto con el operador de selección `[...]` como con los métodos `loc` e `iloc`.

Para seleccionar cierta columna de un _dataframe_ podemos utilizar directamente el nombre de variable que representa, entrecomillado, dentro de los operadores de selección `[]`, y aplicados al *dataframe*, **`dataframe['nombre_variable']`**.

Otra opción es escribir el nombre del dataframe, y aplicar a continuación el punto y el nombre de la variable, **`dataframe.nombre_variable`**.

Si se trata de seleccionar varias columnas, las especificamos seleccionando sobre el dataframe una lista con sus nombres entrecomillados, **`dataframe[["var1","var2",...]]`**.

Existe sin embargo un método más genérico para seleccionar filas y columnas, a través de dos funciones específicas como son:
- `iloc[]` para seleccionar filas y columnas utilizando los numerales que las identifican
- `loc[]` para seleccionar filas y columnas utilizando los nombres o etiquetas que las identifican.

En ambas, como argumentos podemos pasar una etiqueta única, una lista de etiquetas, una porción de etiquetas o una serie booleana (de datos `True/False`).

Estos métodos se aplican a un data frame seguidos de un punto, y con la selección entre corchetes:
```
dataframe.loc[seleccion]
dataframe.iloc[seleccion]
```

Las reglas de selección con estas funciones son las siguientes:
- seleccionar toda una fila *f*, `.iloc[f] / .loc[f]`,
- seleccionar toda una columna *c*, `.iloc[:,c] / .loc[:,c]`
- realizar múltiples selecciones de filas y columnas, `.iloc[lista_f,lista_c] / .loc[lista_f,lista_c]`.

Veamos algunos ejemplos, primero seleccionando por los numerales con `iloc` y luego con las etiquetas con `loc`.

````python
# Selección por numerales

# Selección de la primera fila
print("\n Primera fila:\n",encuesta.iloc[0])
# selección de la última fila
print("\n Última fila:\n",encuesta.iloc[-1])

# Selección de la primera columna
print("\n Primera columna:\n",encuesta.iloc[:,0])
# selección de la última columna
print("\n Última columna:\n",encuesta.iloc[:,-1])
````

````python
# Selección de varias filas y columnas
print("\n Primeras dos filas y primeras dos columnas:\n",encuesta.iloc[0:2,0:2])

# selección de la primera, tercera y cuarta filas (todas las variables)
print("\n primera, tercera y cuarta filas:\n",encuesta.iloc[[0,2,3]])
print("\n primera, tercera y cuarta filas:\n",encuesta.iloc[[0,2,3],:])

# selección de la primera y tercera columnas
print("\n primera, tercera y cuarta columnas:\n",encuesta.iloc[:,[0,2]])
````

## <font color='steelblue'>3. Valores perdidos y detección

Con frecuencia, cuando leemos o incluso creamos una base de datos, podemos encontrar registros en los que no disponemos de información, bien porque no se ha observado/medido, bien porque se ha producido un error en la transcripción y simplemente no se registró. Hablamos aquí sobre  los valores perdidos en una base de datos, la relevancia que pueden tener en un análisis y cómo detectarlos.


Cuando pedimos información sobre la base de datos `encuesta`, ya comprobamos que las  variables _talla_ y _peso_ tenían solo 1777 registros, frente a las 1814 filas del fichero. Es decir, contenían valores perdidos, también llamados valores faltantes. Estos valores se identifican como `NaN` o `nan` en _Python_, que en inglés significa *Not A Number*, y se obtienen con `NaN` o con `np.nan`, del módulo  _NumPy_.

Cuando utilizamos el comando de *Pandas* `read_csv` para leer bases de datos, esta función detecta automáticamente los valores perdidos y los reemplaza con este valor especial `NaN`.

Los valores faltantes en una base de datos pueden llegar a ser relevantes y ocasionar sesgos importantes en las conclusiones, si están vinculados a muchos registros, o a registros con ciertas características. Es pues SIEMPRE muy importante, identificarlos al comenzar a trabajar con un conjunto de datos.

En ocasiones, los investigadores codifican de un modo concreto los valores perdidos (por ejemplo, es común encontrar el valor -1, cuando todos los valores han de ser necesariamente positivos en una variable). Si este es el caso, antes de iniciar el trabajo con la base de datos, habremos de modificar estos valores asignados a los registros perdidos, y asignar el valor `NaN` para poderlo trabajar convenientemente en Python como valor perdido, y excluir por tanto dichos registros en algunos análisis (por ejemplo, no tiene sentido integrar registros `NaN` para calcular un promedio).

Una vez reconocidos adecuadamente los valores faltantes en una base de datos, las estrategias básicas para el tratamiento de los datos son:

1. eliminar todos los registros con datos faltantes, y utilizar exclusivamente aquellos registros que contienen información completa en todas las variables;
2. otra opción, si hay muchos valores faltantes e información suficiente, es imputar valores para no desperdiciar tanta información excluyendo los registros perdidos.

Si bien no trabajaremos en este cuaderno estas estrategias (lo haremos más adelante), sí que presentamos a continuación las funciones que nos ayudarán a detectar valores perdidos
en una base de datos.



### <font color='steelblue'>Funciones para detectar valores perdidos

En los procedimientos básicos del análisis estadístico, *Python* omite este tipo de valores, pero en algunos casos más complejos es conveniente poder identificarlos para trabajarlos de forma adecuada.

Para detectar los valores perdidos, podemos utilizar las siguientes funciones:

* **`isnull()`**: proporciona valor `True` cuando una observación es faltante y `False` para las no faltantes;
* **`notnull()`**: procede al revés de la anterior, e identifica con `True` los valores que no son faltantes;
* **`dropna()`**: filtra los datos eliminando datos faltantes según el criterio que utilicemos como argumento; `how=any` excluya las filas en las que encuentra algún valor faltante en alguna columna; `how=all` excluye solo las filas en las que todas las columnas tienen valores faltantes; `subset='var1'` excluye solo las filas en las que la variable `var1` tiene valores faltantes.


Aplicamos a continuación estas funciones a la base de datos 'encuesta'.  

Al aplicar la función `info()`, observamos la presencia de valores faltantes en las variables `peso` y `talla`, con 1777 registros no nulos cada una, a diferencia de 1814 que es el número total de registros existentes.


````python
# Información básica sobre variables con valores faltantes
encuesta.info()
````

Si queremos depurar la base de datos excluyendo todas las filas en las que hay algún valor faltante, podemos utilizar la selección anterior, o bien directamente la función `dropna(how=any)`, sin tener que utilizar las columnas en la que hay valores faltantes.

Veamos cómo reproduce el resultado anterior.

````python
# Base de datos filtrada con registros completos
encuesta.dropna(how='any')
````

En este caso el argumento `how=all` no modificará la base de datos original, puesto que solo existen valores faltantes en dos de sus columnas: no hay filas completas con valores faltantes.

````python
# Base de datos filtrada excluyendo filas enteras perdidas
encuesta.dropna(how='all')
````

Si se diera el caso de que los valores faltantes en varias columnas están ubicados en filas distintas, cabría filtrar la base de datos con los valores completos para alguna de ellas, y no tomar en consideración los valores faltantes de las otras columnas. Esto lo resolvemos con el argumento `subset`, en el que indicamos las variables a tener en cuenta para identificar los valores perdidos. Esto nos permite plantear todas las posibilidades de filtrado de faltantes incorporando también el argumento `how`.

````python
# Filtramos los registros con datos completos en la variable 'peso'
encuesta.dropna(subset = ['peso'], how = 'any')
````

````python
# Filtramos los registros con datos completos,
# en las variables 'peso' o 'talla'
encuesta.dropna(subset = ['peso', 'talla'], how = 'any')
````

Por último y a modo de curiosidad, pues más adelante aprenderemos funciones de imputación automática de valores perdidos, veamos cómo se pueden rellenar los datos faltantes.
Para ello vamos a utilizar la función `fillna(value=...)`, que se aplica a una columna, indicando en el argumento `value` el valor a imputar, esto es, con el que sustituir los valores faltantes. Para mantener la asignación sobre el dataframe original hay que añadir el argumento `inplace=True`, pues por defecto no se aplican los cambios sobre él; habría que asignar el cambio a un nuevo objeto.

Como ejemplo, vamos a sustituir los valores perdidos de la variable `peso` por el valor mínimo, y comprobar cómo cambia la media, realizando un duplicado del dataframe original.

````python
# Rellenamos los valores perdidos de 'peso' con su mínimo
print('La media con los datos faltantes es',encuesta["peso"].mean())
# creamos un duplicado
encuesta_fill=encuesta
# calculamos el mínimo peso
minimo = encuesta['peso'].min()
# y rellenamos los faltantes, asignando el mínimo
encuesta_fill['peso'] = encuesta_fill["peso"].fillna(value=minimo)
# comprobamos el cambio de media
print('La media con los datos faltantes rellenados es',encuesta_fill["peso"].mean())
````

## <font color='steelblue'> 5. Crear y codificar variables

La creación de nuevas variables es necesaria en ocasiones para identificar patrones y tendencias en conjuntos de datos existentes y derivar información adicional que pueda ser útil para el análisis. Se pueden generar variables que aporten valor adicional a los modelos o estudios realizados, permitiendo obtener una visión más completa y enriquecida de los datos, ya sea a base de combinar información en diversas variables, bien a base de categorizar o incluso recodificar variables ya existentes. Estas son las premisas que estudiamos en esta sección.


En el cuaderno anterior vimos cómo crear variables nuevas asignando valores con una lista (`data_frame["nueva_variable"] = [1, 2, 3, 4]`) o utilizando operadores sencillos sobre algunas ya existentes (`data_frame["nueva_variable"] = data_frame["variable1"] + data_frame["variable2"]`).

Si la base de datos con la que estamos trabajando tiene muchos registros (filas), es difícil crear una variable pasando una lista. Además, si los cálculos implican a más de dos o tres variables, e incluso cálculos complejos, trabajar con los operadores convencionales y la asignación del resultado en una nueva columna puede ser un método un tanto farragoso. Por todo ello se propone una nueva estrategia consistente. en programar funciones específicas que calculan y aplican los valores correspondientes para crear nuevas columnas en el dataframe.


A modo de ejemplo vamos a crear la función `bmi`, que calcula el *índice de masa corporal* (BMI) a partir de la `talla` y el `peso` en el dataframe *encuesta*. Puesto que involucra cálculos con más de una variable de la base de datos, necesitamos de otra función que la aplique a las variables de dicha base de datos; esta función es **`vectorize()`** de _NumPy_. En el argumento de `vectorize` indicamos la función a aplicar y fuera, entre paréntesis `("var1","var2",...)`, las variables a las que se aplica.

Para poder calcular este índice, puesto que la talla viene dada en centímetros, hemos de convertirlos a metros:

$$bmi = \frac{peso (kg)}{talla (m) ^2}.$$

````python
# Determinación del bmi

# Primero convertimos la 'talla' a metros
encuesta["talla"] = encuesta["talla"]/100

# Definimos la función 'bmi'
def bmi(x, y):
  '''
  Calcula la división del primer argumento entre el cuadrado del segundo

  Argumentos:
  - x: objeto de Python que sea operable
  - y: objeto de Python que sea operable

  Devuelve:
    el valor de 'x' entre 'y' al cuadrado, redondeado a las centésimas
  '''
  return round(x/y**2, 2)

# Aplicamos la función 'bmi' a las variables 'peso' y 'talla', para crear la variable 'bmi'
encuesta["bmi"] = np.vectorize(bmi)(encuesta["peso"], encuesta["talla"])

# Mostramos las 5 primeras filas
encuesta.head()
````

La función **`vectorize()`** se utiliza cuando la función se aplica a más de una  variable de la base de datos. Si queremos construir una nueva variable aplicando una función a una única variable, utilizamos la función **`apply()`** de _Pandas_.

Para ejemplificar su uso, una vez construida la variable `bmi`, queremos utilizarla para clasificar a los sujetos en la categorización que propone la OMS:

<center><img src=https://raw.githubusercontent.com/ia4legos/Statistics/main/images/img_170_20_10.png width="350" height="200"></center>

Definimos a continuación la función `clasificar_bmi()`, para clasificar a los sujetos según su `bmi` en dichos niveles, y guardamos el resultado en una nueva variable llamada `bmi_agrupado`.

````python
# Categorización de variables numéricas con función propia
# clasificación OMS de los sujetos por su bmi
def clasificar_bmi(x):
  '''
  Clasifica a los individuos en grupos dependiendo de su bmi

  Argumentos:
    - x: número

  Devuelve:
    - si el valor de 'x' es menor que 18.5, Infrapeso
    - si el valor de 'x' está entre 18.5 (inclusive) y 25, Normal
    - si el valor de 'x' está entre 25 (inclusice) y 30, Sobrepeso
    - si el valor de 'x' está entre 30 (inclusive) y 35, Obesidad1
    - si el valor de 'x' está entre 35 (inclusive) y 40, Obesidad2
    - si el valor de 'x' es mayor o igual a 40 (o en caso contrario), Obesidad3
  '''
  if x<18.5:
    return "Infrapeso"
  elif 18.5<=x<25:
    return "Normal"
  elif 25<=x<30:
    return "Sobrepeso"
  elif 30<=x<35:
    return "Obesidad1"
  elif 35<=x<40:
    return "Obesidad2"
  else:
    return "Obesidad3"

encuesta["bmi_agrupado"] = encuesta["bmi"].apply(clasificar_bmi)
encuesta.head()
````

Hemos hecho lo que se denomina **categorización de una variable numérica por niveles de respuesta**. Esto lo podemos hacer automáticamente con la función **`cut()`** de _Pandas_, sin necesidad de programar una nueva función. Los argumentos de la función `cut()` son: la variable a categorizar, los límites de cada nivel (`bins`), las etiquetas de cada nivel (`labels`). Si queremos que los intervalos de respuesta estén abiertos por la derecha, hemos de indicar en los argumentos `right=False`, pues por defecto crea intervalos cerrados por la derecha (`límite inferior < x <= límite superior`) y nos interesa lo contrario (`límite inferior <= x < límite superior`).

Para aplicar la función `cut()` eliminamos la que hemos creado antes con la categorización, `bmi_agrupado` (con `drop()`), y utilizamos el mismo nombre para volcar el resultado.

````python
# Categorización de variables numéricas con cut()

# eliminamos la variable 'bmi_agrupado' para un nuevo dataframe
encuesta2 = encuesta.drop(columns='bmi_agrupado')

# volvemos a crear 'bmi_agrupado' con cut()
encuesta2["bmi_agrupado"] = pd.cut(encuesta2["bmi"],
                                   bins = [0, 18.5, 25, 30, 35, 40, encuesta2["bmi"].max()],
                                   labels = ["Infrapeso", "Normal", "Sobrepeso", "Obesidad1", "Obesidad2", "Obesidad3"],
                                   right=False)
encuesta2.head()
````

Sobre una variable categórica, en determinados contextos nos puede interesar recodificar los valores de respuesta, esto es, asignar otras etiquetas a los niveles de respuesta. Esto se llama **recodificación de variables categóricas**. Para esta tarea podemos utilizar varias funciones, como **`replace()`** o **`map()`**.
```
dataframe.replace({'var1':'var1_nueva','var2':'var2_nueva',...})
```

Seguimos sobre el mismo ejemplo. Dado que la variable `bmi_agrupado` tiene varios niveles de Obesidad (1,2,3), si queremos agruparlos todos ellos en un único nivel, esto es, **colapsarlos**, aplicamos los cambios de etiquetas (Obesidad1=Obesidad, Obesidad2=Obesidad y Obesidad3=Obesidad) a través de la función `replace()` y un diccionario con los cambios. Aplicamos los cambios sobre la misma variable `bmi_agrupado` con el argumento `inplace=True`.


````python
# Recodificar una variable: colapsar niveles

# los niveles de respuesta original
print(encuesta.bmi_agrupado)
# colapsamos todos los niveles de Obesidad1,Obesidad2 y Obesidad3 en uno solo: Obesidad
encuesta.replace({"Obesidad1": "Obesidad", "Obesidad2": "Obesidad", "Obesidad3": "Obesidad"}, inplace=True)
print(encuesta.bmi_agrupado)
````
