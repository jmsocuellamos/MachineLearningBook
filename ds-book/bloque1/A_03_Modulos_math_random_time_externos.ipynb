## <font color='steelblue'> 1. ¿Qué es un módulo?


Un **módulo** en Pyhton es un fichero de extensión *.py* que contiene funciones que pueden importarse en otros programas (cuadernos Jupyter).

Cuando nuestro programa se hace largo, podemos dividirlo en varios archivos (módulos) lo que nos facilita su mantenimiento. Los módulos también nos permiten utilizar una función sin tener que escribir el código en cada programa (cuaderno Jupyter).

Por ejemplo, podemos usar un editor de texto para crear un fichero *'fibo.py'* donde definimos dos funciones:

```
# Fibonacci numbers module
def fib(n): # write Fibonacci series up to n
	a, b = 0, 1
	while a < n:
		print(a, end=' ')
		a, b = b, a+b
	print()

def fib2(n): # return Fibonacci series up to n
	result = []
	a, b = 0, 1
	while a < n:
		result.append(a)
		a, b = b, a+b
	return result
```

Subimos este fichero 'fibo.py' a nuestros Archivos del cuaderno Jupyter y podremos usar esas dos funciones con una sola línea de´codigo:

```
import fibo
```
Solo tenemos que usar el nombre del módulo y llamar a la función que queramos:


````python
import fibo
````

````python
fibo.fib(1000)
fibo.fib2(100)
````

En Python se definen por defecto una serie de **módulos básicos internos**, como los **módulos `math`, `random` y `time`** que vamos a estudiar en este cuaderno, y que proporcionan un conjunto de funciones muy utiles para cualquier programador. Podemos consultar los módulos internos en la Documentación Oficial en  https://docs.python.org/3/py-modindex.html.


Puesto que la comunidad de Python está en continua evolución, se va enriqueciendo con las aportaciones de los usuarios. Todos los módulos que se programan cuentan con referencias y ayudas para hacerlos accesibles al resto de usuarios.
Los módulos externos (en *PyPI*) necesitaremos instalaros en nuestro entorno de trabajo, puesto que no vienen preinstalados con Python como los módulos internos.

 Una vez instalados los módulos externos, necesitaremos  **importarlos**. En este cuaderno veremos cómo  utilizar y consultar las funcionalidades de los módulos de Python, tanto si son internos como externos.

## <font color='steelblue'> 2. Módulo math


El **módulo `math`** proporciona acceso a las funciones matemáticas definidas en el estándar de C. La documentación completa de la librería se puede consultar en https://docs.python.org/es/3/library/math.html.

Utilizamos este módulo para ilustrar cómo utilizar cualquier función contenida en él. Veremos a continuación algunas funciones relevantes de dicho módulo que se utilizan con frecuencia.

### <font color='steelblue'> Importar un módulo interno

Para utilizar una función de un módulo dado llamado `mod`, necesitamos primero que el módulo esté disponible en nuestro programa. Para ello hemos de 'importarlo' con la **instrucción `import`**.

```
import mod
```


Por ejemplo, para calcular la raíz cuadrada de un número queremos utilizar una función llamada `sqrt` que está dentro del módulo `math`. Importamos pues, dicho módulo con la sintaxis:

````python
import math
````

Para saber cuáles son las funciones que contiene un módulo podemos consultar directamente su tutorial en la web, o solicitar en nuestro programa un listado con la función `dir()`.

La **función dir(objeto)** devuelve el conjunto de nombres asociados al objeto incluido como argumento. Si no se incluye ningún argumento, devuelve el conjunto de nombres del ámbito local.

````python
dir(math)
````

Una vez importado un módulo, podemos llamar a la función que necesitamos, utilizando como prefijo el nombre del módulo que la contiene, seguido de un punto.

Así si queremos, por ejemplo, calcular la raíz cuadrada de un número, lo definimos y a continuación llamamos a la función con la sintaxis:




````python
x = 5
x = math.sqrt(x)
print(x)
````

Al poner delante de `sqrt` el módulo `math` y un punto, estamos usando lo que se denomina un *'espacio de nombres'*, que permite especificar exactamente qué función `sqrt` queremos utilizar, dado que podría estar programada con el mismo nombre en otro módulo: en nuestro caso queremos usar la función `sqrt` que está programada en el módulo `math`.



Como ejemplo de buena praxis se recomienda colocar todas las declaraciones `import` de todos los módulos que vamos a utilizar en un programa, en la parte superior del archivo de programa (o en la parte superior de la primera celda cuando se usa un cuaderno Jupyter). Así podremos cargar en bloque todos los módulos necesarios al inicio para tenerlos accesibles en cualquier momento.

### <font color='steelblue'> Espacios de nombres

Una función de un módulo viene identificada unívocamente con su **espacio de nombres**, que especifica el módulo que la contiene, seguido de un punto y del nombre de la función.

Utilizar este prefijo (el módulo que la contiene seguido de un punto) puede parecer redundante, pero en la práctica a menudo hay diferentes algoritmos con el mismo nombre para realizar la misma operación o una similar, y que pueden variar en velocidad y precisión. Si no se distinguen por su *espacio de nombres*, se producirá un *'conflicto' de nombres*. Los espacios de nombres nos permiten pues, identificar exactamente qué algoritmo queremos utilizar, y son extremadamente útiles cuando la generación de código proviene de muchos autores, como ocurre en *Python*.


### <font color='steelblue'> Importar una función

Si solo estamos interesados en utilizar una función específica de un módulo, en lugar de importar el módulo completo podemos importarla directamente, de manera que ya no tendremos que recurrir a su espacio de nombres para llamarla. Esto se hace con la sintaxis:

``from mod import fun``

donde 'mod' se refiere al módulo en el que está contenida la función `fun`.

El mismo ejemplo anterior para calcular la raíz cuadrada lo resolvemos directamente llamando a la función `sqrt` con:

````python
from math import sqrt

x = 5
x = sqrt(x)
print(x)
````

### <font color='steelblue'> Renombrar una función

En ocasiones nos puede interesar renombrar algunas funciones que importemos, con el fin de utilizar nuestra propia nomenclatura. Esta cuestión la podemos resolver importando la función de interés como vimos antes, y atribuyéndole el nombre que deseemos con el comando `as`:

``from mod import fun as mi_nombre``

A continuación, para utilizar dicha función bastará llamar a `mi_nombre()` con los argumentos pertinentes de la función `fun()`.

Por ejemplo, si en nuestro programa queremos renombrar la función `sqrt` como `raiz_cuadrada`, escribiremos:

````python
from math import sqrt as raiz_cuadrada

x = 5
x = raiz_cuadrada(x)
print(x)
````

### <font color='steelblue'> Redondear

El módulo `math` nos proporciona las **funciones `floor()`, `ceil()` y `trunc()`**, que redondean los números decimales al entero por defecto o el primer entero menor que el número (`floor`), por exceso o el primer entero mayor que el número (`ceil`), o simplemente eliminan la parte decimal dejando la parte entera del número (`trunc`).

````python
# Entero por defecto

x = 5.2346
y = 6.87567
print(math.floor(x))
print(math.floor(y))
````

````python
# Entero por exceso

x = 5.2346
y = 6.87567
print(math.ceil(x))
print(math.ceil(y))
````

````python
# Trunc

x = 5.2346
y = 6.87567
print(math.trunc(x))
print(math.trunc(y))
````

### <font color='steelblue'> Factoriales

El factorial de un número es la cantidad que resulta de multiplicar un determinado número natural por todos sus antecesores excluyendo el 0, y que se expresa matemáticamente con el símbolo `!`. Por ejemplo, el factorial de 4 sería:

$$ 4! = 4 \cdot 3 \cdot 2 \cdot 1 = 24 $$

El módulo `math` nos facilita el cálculo de los factoriales con la **función `factorial()`**:

````python
x = 4
print(math.factorial(x))
````

### <font color='steelblue'> Valores especiales

Existen algunas funciones que nos permiten comprobar si un objeto tiene valores especiales, como los valores indefinidos o no representables `NaN` o valores infinito $\infty$. En el módulo `math` tenemos la **función `isnan()`** para identificar valores indefinidos, y la **función `isinf()`** para identificar valores infinitos.

Así `math.isnan(x)` devolverá `True` si `x == math.nan`, y `False` si `x != math.nan`.

Veamos un código para comprobar si un número (introducido previamente) es indefinido o no: si es indefinido, mostramos un mensaje que informa sobre ello, si no lo es, muestra el número.

````python
x = 5
if math.isnan(x):
  print("Valor indefinido: {math.isnan(x)}")
else:
  print(f'Valor asignado: {x}')
````

Utilizamos también la función `isinf()`, para ver lo que devuelve cuando asignamos a un objeto `x` el valor `infinito`.

````python
# isinf
x = math.inf
print(f'Valor infinito: {math.isinf(x)}')
````

## <font color='steelblue'> 3. Módulo random

El módulo `random` implementa generadores de números pseudo-aleatorios para varias distribuciones comunes. La documentación completa se puede consultar en la Documentación Oficial de Pyton:  http://docs.python.org/3/library/random.html. Veremos varias funciones de este módulo que nos pueden resultar especialmente útiles.

En primer lugar, para acceder a las funciones de este módulo, debemos importarlo.

````python
import random
````

````python
dir(random)
````

Son muy útiles en este módulo las funciones `choice()`, `random()`, `uniform()` y `shuffle()`. Después de llamar a `import random` podemos acceder a estas funciones mediante `random.choice()`, `random.random()`, `random.uniform()` y `random.shuffle()`. También podemos importarlas directamente para no tener que copiar el nombre del módulo, con las palabras reservadas (keywords) `from ... import ...`.

Veremos también las semillas aleatorias, que son útiles para reproducir un mismo resultado de simulación desde una misma máquina.

### <font color='steelblue'> Función random(): Aleatoriedad entre 0 y 1

La **función `random()`** genera un número aleatorio decimal entre 0 y 1, esto es, proveniente de una distribución uniforme $Un(0,1)$:

````python
print(random.random())
````

Por ejemplo, si queremos simular 5 números aleatorios entre 0 y 1 podemos llamar a esta función a través de un bucle y almacenarlos en una lista vacía que habremos definido previamente, usando la función `append()`; si además solo queremos conservar 3 cifras decimales, habremos de utilizar la función `round()`:

````python
lista = []
for i in range(10):
    lista.append(round(random.random(),3))
print(lista)
````

### <font color='steelblue'> Función seed(): Semillas aleatorias

Al llegar aquí te habrás dado cuenta de que si lanzas dos veces seguidas un simulador, obtienes siempre valores distintos.

Es interesante a la hora de simular, tener en cuenta que todos los simuladores programados son simuladores pseudo-aleatorios, esto es, no completamente aleatorios, puesto que arrancan siempre de una semilla. Así pues, si un simulador en un ordenador utiliza siempre una misma semilla, simulará siempre los mismos valores. Cuando queramos reproducir siempre los mismos resultados desde una misma máquina, bastará pues que prefijemos la semilla antes de ponernos a simular, con la **función `random.seed()**`.

Veámoslo en acción. Inicializamos semilla y lanzamos el simulador. Volvemos a inicializar la misma semilla y volvemos a lanzar el simulador. Vemos cómo conseguimos la misma simulación.

````python
random.seed(1234)
print(f'Primera simulación {random.random()}')
random.seed(1234)
print(f'Segunda simulación {random.random()}')
````

````python
random.seed(1234)
print(f'Primera simulación {random.random()}')
print(f'Segunda simulación {random.random()}')
````

Cuando ejecutemos el código en otra máquina con la misma semilla, lo más seguro es que no consigamos el mismo resultado que en la primera máquina, pues el simulador utiliza otros parámetros de reloj interno del computador.

### <font color='steelblue'> Función uniform(): Aleatoriedad en intervalo

La **función `uniform()`** nos proporciona un número decimal entre dos valores dados `a` y `b`; esto es, simula de una distribución $Un(a,b)$. Esta función opera igual si ponemos antes o después el valor mayor o el menor.

Queremos simular, por ejemplo, un número aleatorio entre el 1 y el 9:

````python
print(random.uniform(1,9))
````

````python
print(random.uniform(9,1))
````

### <font color='steelblue'>Función choice(): Elegir

La **función `choice()`** nos permite seleccionar al azar un elemento de una secuencia, esté en una lista o en una tupla. Simularía por ejemplo, una tirada de un dado:

````python
lista = list(range(6))
print(lista)
print(random.choice(lista))
````

````python
tupla = tuple(range(6))
print(tupla)
print(random.choice(tupla))
````

### <font color='steelblue'> Función shuffle(): Barajar

La **función `shuffle()`** nos permite permutar una lista, es decir, cambiar aleatoriamente el orden los valores que contiene. Esta función, aplicada a una lista, la modifica, de modo que cuando la volvamos a imprimir la veremos barajada.


````python
print(f'Lista original: {lista}')
random.shuffle(lista)
print(f'Lista barajada: {lista}')
````

## <font color='steelblue'> 4. Módulo time

El **módulo `time`** contiene distintas funciones relacionadas con el tiempo y es muy útil, por ejemplo, para cronometrar tiempos de ejecución de un código. La documentación completa se puede consultar en  http://docs.python.org/3/library/time.html.



### <font color='steelblue'> Función time(): Tiempo transcurrido

La **función `time()`** devuelve el tiempo en segundos desde la época origen (January 1, 1970, 00:00:00 (UTC)), como un número en coma flotante.
La época podemos obtenerla ejecutando `time.gmtime(0)` que nos devuelve una estructura tiempo:


````python
# Fecha de referencia para calcular el tiempo
import time
print(time.gmtime(0))
print()
print(time.time())
````

Para no confundir la función `time()`con el módulo `time`, podemos a renombrarla como `tiempo` y así obtener la ayuda para esa función:

````python
# importamos la función 'time' en el módulo 'time' y la renombramos a 'tiempo'
from time import time as tiempo
help(tiempo)
````

La **función `gmtime([secs])`** convierte un tiempo expresado en segundos desde la época (=January 1, 1970, 00:00:00 (UTC)) a un `struct_time` en UTC en el que el flag `dst` es siempre cero. Si no se proporciona 'secs', se utiliza la hora actual devuelta por time(). Se ignoran las fracciones de segundo.

````python
ahora = time.gmtime()
print(ahora)
````

Si queremos contabilizar el tiempo de ejecución de un proceso, bastará guardar el valor del tiempo antes de empezarlo y volverlo a registrar después, para calcular la diferencia.

Veamos un ejemplo: hacemos 10 tiradas aleatorias de un dado con una función del módulo `random`.



````python
import random

time_1 = tiempo()
dado=list(range(6))
tiradas=range(10)
for i in tiradas:
  random.choice(dado)
time_2 = tiempo()
print(f"Iniciamos el programa en t={round(time_1/(3600*24*365),2)} años después del 01/01/1970")
print(f"Terminamos las tiradas t={round(time_2,1)} segundos después del 01/01/1970")
print(f'Tardamos {time_2-time_1} segundos en hacer las {len(tiradas)} tiradas.')
````

### <font color='steelblue'>Función ctime(): Fecha actual

La **función `ctime()`** proporciona el instante actual en la fecha y hora actual en el meridiano de Greenwich o GMT (Greenwich Mean Time).

````python
import time
fecha_hora_local = time.ctime(tiempo())
print(f"La fecha y hora local es: {fecha_hora_local}")
````

Sabiendo que la referencia es la fecha/hora GMT, dependiendo de si estamos en horario de verano o invierno (DST = Daylight Saving Time), la desviación en España será de dos o una hora más respectivamente. Para los cálculos hemos de transformar a segundos.

````python
fecha_hora_local = time.ctime(tiempo())
print(f"La fecha y hora local GMT es: {fecha_hora_local}")
fecha_hora_local = time.ctime(tiempo()+2*3600)
print(f"La fecha y hora local en verano es: {fecha_hora_local}")
fecha_hora_local = time.ctime(tiempo()+1*3600)
print(f"La fecha y hora local en invierno es: {fecha_hora_local}")
````

### <font color='steelblue'>Función gmtime(): Fechas desglosadas

Si bien ya la utilizamos antes para comprobar cuál es la fecha desde la que se contabiliza el tiempo con la función `time()`, tenemos otra aplicación posible y muy útil para la **función `gmtime()`**: generar archivos o carpetas cuyo nombre contenga la fecha/hora de creación de un proyecto.

Por ejemplo, queremos que nuestro archivo o carpeta reciba por nombre los datos correspondientes al año, mes, día y hora en que hemos generado este proyecto, denominado 'xyz', es decir, queremos que se llame `año_mes_dia_hora_proyecto_xyz`.

Para recopilar toda esta información y poderla concatenar a nuestro gusto, tenemos la función `gmtime()`, que devuelve una tupla con toda la información horaria del instante que le indiquemos.


````python
help(time.gmtime)
````

Los resultados que se obtienen con `gmtime()` empiezan por `tm_` y nos propocionan la información temporal que queremos de la fecha solicitada. Así `tm_year` da el año o `tm_hour` la hora.

Vamos a generar a continuación una cadena de caracteres (*string*) para dar nombre a nuestro proyecto.

````python
time_actual = tiempo()
fecha = time.gmtime(time_actual+2*3600)
print(fecha)
anyo = fecha.tm_year
mes = fecha.tm_mon
dia = fecha.tm_mday
hora = fecha.tm_hour
minutos = fecha.tm_min
nombre_carpeta = str(anyo)+"_"+str(mes)+"_"+str(dia)+"_"+str(hora)+"_"+str(minutos)+"_proyecto_xyz"
print(f'El nombre de la carpeta es:{nombre_carpeta}')
````

Otras que no hemos usado son `tm_wday` o día de la semana (siendo 0 el lunes), `tm_yday` o día del año.

## <font color='steelblue'> 5. Módulos externos

### <font color='steelblue'> Instalación e importación

Existen algunos de módulos que no se encuentran en las bibliotecas de *Python* como módulos internos, y que tampoco están preinstalados en nuestro entorno, por lo que requieren un paso previo a la importación para su utilización: la **instalación**. Es más, algunos módulos están incluidos en paquetes de módulos y será preciso instalar los paquetes para poder acceder al módulo que nos interesa y a sus funciones y elementos. Los pasos a dar serán:

1. Instalación del paquete/módulo contenedor.
1. Importación del módulo de interés (o funciones).



Para instalar un paquete, *Python* dispone del **paquete `pip`** (también es una palabra reservada). El gestor de paquetes `pip` nos permitirá instalar, actualizar y eliminar paquetes del repositorio PyPI (https://pypi.org/) o de otros repositorios. Algunos comandos pip interesantes son:


*   **pip list**: muestra los paquetes instalados y su versión
*   **pip list --outdated**: muestra los paquetes obsoletos
* **pip install nombre_paquete**: instala un paquete
* **pip install --upgrade nombre_paquete**: actualiza un paquete a la última versión
* **pip uninstall nombre_paquete**: desinstala un paquete
* **pip check**: comprueba que los paquetes instalados no tienen problemas de dependencias con otros paquetes

Para más información consultar en https://pypi.org/project/pip/

Para ejecutar comandos externos a Python en un cuaderno se utiliza una sintaxis especial:
```
!pip parámetro(s)
```

El signo de exclamación (`!`) y la palabra `pip` no tienen por qué ir pegados.




````python
# listado de paquetes instalados en mi cuaderno
! pip list
````
