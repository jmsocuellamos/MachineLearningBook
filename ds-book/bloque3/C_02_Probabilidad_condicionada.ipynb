## <font color='steelblue'>Introducción</font>

**Descripción:** La probabilidad condicionada se refiere a la probabilidad de que ocurra un evento A, dado que otro evento B ha ocurrido. Nos permite analizar cómo la ocurrencia de un evento afecta la probabilidad de otro evento relacionado.

La regla de multiplicación es una herramienta fundamental en la teoría de la probabilidad. Nos permite calcular la probabilidad conjunta de dos eventos A y B al multiplicar la probabilidad de A dado B por la probabilidad de B. Es aplicable cuando los eventos son independientes o cuando se cumple la probabilidad condicionada.

La independencia es un concepto clave en la teoría de la probabilidad. Dos eventos A y B se consideran independientes si la ocurrencia de uno no afecta la probabilidad del otro. Esto implica que la probabilidad conjunta de A y B es igual al producto de las probabilidades individuales de A y B.

La ley de probabilidad total se utiliza para calcular la probabilidad de un evento A mediante la consideración de diferentes casos o subconjuntos que cubren todos los posibles resultados. Se basa en la idea de que la probabilidad de A es la suma de las probabilidades condicionadas de A dado cada subconjunto.

El teorema de Bayes es una herramienta poderosa para actualizar nuestras creencias o probabilidades sobre un evento dado la evidencia nueva. Nos permite calcular la probabilidad condicionada de un evento A dado un evento B, utilizando la probabilidad condicionada de B dado A y las probabilidades iniciales de A y B.

La independencia condicionada se refiere a la independencia entre dos eventos A y B, dado un tercer evento C. Esto significa que, aunque los eventos A y B pueden no ser independientes por sí solos, se vuelven independientes cuando se condicionan en la ocurrencia de C.

## <font color='steelblue'> 1. Probabilidad condicionada

En este apartado se van a tratar conceptos más fundamentales de la teoría de la probabilidad. Nos vamos a centrar en que, una vez obtenemos nueva información sobre sucesos, cómo vamos a poder actualizar las probabilidades que tenemos. Imaginemos que, en una ciudad determinada, llueve el 37% de los días. Con esto sabemos que, si elegimos un día al azar en esta ciudad, tenemos un 37% de probabilidades de que llueva ese día. Expresado de otra manera: $P(LL) = 0.37$, siendo $LL$ el suceso de que llueva el día que hemos escogido al azar. Ahora bien, además de seleccionar un día al azar, también sabemos que ese día está nublado. Con esta información adicional, ¿de qué manera podríamos actualizar nuestra probabilidad de que ese día llueva? En otras palabras, ¿qué probabilidad hay de que llueva en una ciudad si ese día está nublado?

Este tipo de probabilidad se le conoce como **probabilidad condicionada**, cuando queremos saber la probabilidad de que ocurra un suceso dado otro. Volviendo al ejemplo anterior, consideraremos $N$ el suceso de que esté nublado y para indicar la probabilidad que estamos buscando, que llueva sabiendo que está nublado, lo representamos como $P(LL|N)$. Esta es la *probabilidad condicional* de $LL$ dado que $N$ se ha producido. Pongamos un pequeño ejemplo para ver la fórmula correspondiente a esta probabilidad:

Se lanza un dado no trucado de 6 caras y tenemos el suceso $A$, que recoge los números impares ($A=\{1, 3, 5\}$), y el suceso $B$, formado por los resultados mayores que 3 ($B=\{4, 5, 6\}$). Primero vamos a calcular la probabilidad de $A$, $P(A)$. Teniendo el espacio muestral (finito) $S=\{1, 2, 3, 4, 5, 6\}$, por lo que $P(A) = \frac{|A|}{|S|} = \frac{|\{1, 3, 5\}|}{|\{1, 2, 3, 4, 5, 6\}|} = \frac{3}{6} = \frac{1}{2} = 0.5$. Ahora, queremos saber la probabilidad de $A$ dado $B$, $P(A|B)$. Si sabemos que $B$ ha ocurrido, el resultado deberá se $4, 5$ o $6$, entonces para que $A$ también ocurra deberemos fijarnos en $A \cap B$, que en este caso es $\{5\}$. Sabiendo esto, la *probabilidad condicionada* será igual a la *cardinalidad* de la intersección entre ambos sucesos entre la longitud de $B$: $P(A|B) = \frac{|A \cap B|}{|B|} = \frac{|\{5\}|}{|\{4, 5, 6\}|} = \frac{1}{3} = 0.33$. Esta fórmula es así debido a que tenemos que ver todas las posibilidades de que salga un resultado contenido en $A$ sabiendo que antes se ha dado alguno de los de $B$, es decir, de todas las posibilidades de $B$, cuántas podrían corresponder a $A$.

Con este ejemplo, pongamos la expresión formal de la *probabilidad condicionada*:

> Si $A$ y $B$ son dos sucesos en un espacio muestral $S$, la probabilidad de $A$ dado $B$ se define como $P(A|B) = \frac{|A \cap B|}{|B|} = \frac{P(A \cap B)}{P(B)}$, cuando $P(B) > 0$.

Podría calcularse también la probabilidad de $B$ dado $A$ como $P(B|A) = \frac{|A \cap B|}{|A|} = \frac{P(A \cap B)}{P(A)}$, ya que $A \cap B$ y $B \cap A$ son equivalentes.

La condición de que la probabilidad del suceso $B$ debe ser mayor estricto que $0$, es debido a que, si es cero, $B$ nunca se dará y la *probabilidad condicionada* no existirá (no se puede dividir entre $0$), y si es menor, esta probabilidad será negativa y las probabilidades se encuentran entre $0$ y $1$. Veamos ahora la explicación de la fórmula, suponiendo los sucesos $A$ y $B$ del espacio muestral $S$. Sabemos que $B$ sucede, por tanto, se descartarán todos los posibles resultados que estén fuera de este suceso, reduciendo nuestro espacio muestral solo a $B$. Con esto, la única forma de que pueda darse algún resultado de $A$, es en el conjunto de la intersección, $A \cap B$. Entonces, dividimos la probabilidad del suceso de la intersección ($P(A \cap B)$) entre la del conjunto $B$ ($P(B)$).

La *probabilidad condicionada* es una medida de probabilidad en sí, por lo que satisface los siguientes *axiomas* de la teoría de la probabilidad:

* Para cualquier suceso $A$, se da que $P(A|B) \geq 0$.
* La probabilidad del suceso $B$ dado $B$ es 1 $\Rightarrow P(B|B) = 1$.
* Dados los sucesos *disyuntos* $A_1, A_2, A_3, \ldots$, entonces $P(A_1 \cup A_2 \cup A_3 \ldots | B) = P(A_1|B) + P(A_2|B) + P(A_3|B) \ldots$.


Además, todas las reglas de probabilidad que hemos visto hasta ahora, las podemos extender a la *probabilidad condicionada*. Teniendo los sucesos $A$, $B$ y $C$, con $P(C)\gt 0$, podemos sacar que:

* $P(\bar{A}|C) = 1 - P(A|C)$.
* $P(\emptyset|C) = 0$.
* $P(A|C) \leq 1$.
* $P(A - B|C) = P(A|C) - P(A \cap B|C)$.
* $P(A \cup B|C) = P(A|C) + P(B|C) - P(A \cap B|C)$.
* $Si \ A \subset B \Rightarrow P(A|C) \leq P(B|C)$.

Veamos ahora 3 casos especiales de la *probabilidad condicionada*:

1. Cuando $A$ y $B$ son eventos disjuntos, tenemos que $A \cap B = \emptyset$, entonces: $P(A|B) = \frac{P(A \cap B)}{P(B)} = \frac{P(\emptyset)}{P(B)} = \frac{0}{P(B)} = 0$. Esto tiene sentido porque si $A$ y $B$ son disjuntos, no pueden ocurrir a la vez. Entonces si sabemos que se da $B$ primero, la probabilidad de que ocurra $A$ será $0$.

1. Si $B$ es un subconjunto de $A$ ($B \subset A$), siempre que ocurra $B$, también lo hará $A$, ya que los resultados de $B$ se encuentran también en $A$. Por lo que si nos dicen que dado $B$, calcular la probabilidad de que se dé $A$, esta será $1$. Tenemos que $A \cap B = B$, entonces: $P(A|B) = \frac{P(A \cap B)}{P(B)} = \frac{P(B)}{P(B)} = 1$.

1. Cuando $A$ es un subconjunto de $B$ ($A \subset B$), en este caso $A \cap B = A$. Por lo que tenemos que: $P(A|B) = \frac{P(A \cap B)}{P(B)} = \frac{P(A)}{P(B)}$.

Una vez sabemos todo esto, vamos a pasar a aplicar algunos de nuestros nuevos conocimientos con la ayuda del código de **Python**. Trabajaremos de nuevo con las bolsas de tela que contienen bolas, que utilizamos en el cuaderno anterior. Vamos a extraer 3 bolas de la bolsa que contiene una bola verde, una azul y una roja. El orden de extracción se tendrá en cuenta y una vez la bola esté fuera, se vuelve a introducir en la bolsa. Con lo que tendremos permutaciones de un espacio muestral de tamaño 3 ($n=3$), con 3 extracciones ($k=3$).

```python
# Importamos el módulo itertools
import itertools

# Muestra: una bola azul, una roja y una verde
bolsa2 = {"azul", "roja", "verde"}

# Lista vacía que contendrá las tuplas
ordenado_con_reemplazamiento2 = []

# Bucle for para rellenar la lista vacía con las tuplas al hacer permutación con reemplazamiento
for i in itertools.product(bolsa2, repeat=3):
  ordenado_con_reemplazamiento2.append(i)

# Mostramos las tuplas con las posibles disposiciones de las bolas
for j in ordenado_con_reemplazamiento2:
  print(j)
```

Como ya sabíamos, hay 27 variaciones posibles. Ahora bien, realmente lo que queremos saber es la probabilidad de extraer una bola azul en la segunda posición dado sabiendo que la primera que se sacó fue roja. Para esto, deberemos averiguar el número de permutaciones que hay que cumplan con que la primera bola extraída es de color rojo y en cuántas se extrae primero una roja y luego una azul. Para esto nos vamos a ayudar de bucles `for` y condicionales `if` para encontrarlas. Luego las mostraremos e indicaremos el número que hay para cada caso.

```python
# Listas vacías
roja_primero = []
roja_y_azul = []

# Bucles for que recorre las 27 permutaciones
for j in ordenado_con_reemplazamiento2:
  # Si la primera extraída fue roja
  if j[0] == 'roja':
    # se añade en la permutación en la lista roja_primero
    roja_primero.append(j)
    # si, además, la segunda es azul
    if j[1] == "azul":
      # se añade en la permutación en la lista roja_y_azul
      roja_y_azul.append(j)

# Mostramos las variaciones en las que se extrae primero una bola roja
print("Roja primera")
for i in roja_primero:
  print(i)

# Mostramos las variaciones en las que se extrae primero una bola roja y luego una azul
print("Primero roja y segunda azul")
for i in roja_y_azul:
  print(i)

# Resultados posibles de extraer primero una bola roja
print(f"Hay {len(roja_primero)} resultados posibles si la primera bola extraída sea roja.")

# Resultados posibles de extraer primero una bola roja y luego una azul
print(f"Hay {len(roja_y_azul)} resultados posibles de que la primera bola extraída sea roja y la segunda azul.")
```

Tenemos de que el espacio muestral total (las 27 permutaciones con repetición al realizar 3 extracciones de una muestra de tamaño 3), hay 9 posibilidades de extraer primero una bola roja y 3 de seleccionar primero una bola roja y luego una azul. Sabemos también que el suceso $A$ (extraer primero una bola roja y luego una azul) es un subcojunto de $B$ (sacar primero una bola roja). Entonces, para obtener la *probabilidad condicionada* de extraer en segunda posición una bola azul dado que la primera ha sido roja, podemos realizarlo de 2 maneras diferentes. La primera es dividir el número de posibilidades que tenemos de extraer primero una bola roja y luego una azul, entre de cuántas formas se puede sacar primero una bola roja. Es decir, estamos aplicando la división de las *cardinalidades* de cada suceso: sabemos que $A \subset B \Rightarrow A \cap B = A \Rightarrow P(A|B) = \frac{|A \cap B|}{|B|} = \frac{|A|}{|B|}$.

```python
# Calculamos la probabilidad condicionada dividiendo el tamaño de cada uno de los sucesos
P_condicionada1 = len(roja_y_azul)/len(roja_primero)

# Mostramos la probabilidad condicionada en forma de porcentaje con 2 decimales
print(f"La probabilidad condicionada de extraer en segunda posición una bola azul dado que la primera ha sido roja es del {round(P_condicionada1*100, 2)}%")
```

La segunda opción es calcular primero la probabilidad de extraer primero una bola roja y la de extraer primero una bola roja y luego una azul. Una vez tenemos estas dos, dividiremos una entre la otra. Sabiendo de antes que $A \subset B \Rightarrow P(A|B) = \frac{P(A \cap B)}{P(B)} = \frac{P(A)}{P(B)}$.

```python
# Calculamos la probabilidad de extraer primero una bola roja
P_roja = len(roja_primero)/len(ordenado_con_reemplazamiento2)
# Calculamos la probabilidad de extraer primero una bola roja y luego una azul
P_roja_y_azul = len(roja_y_azul)/len(ordenado_con_reemplazamiento2)

# Calculamos la probabilidad condicionada dividiendo la probabilidad de cada uno de los sucesos
P_condicionada2 = P_roja_y_azul/P_roja

# Mostramos la probabilidad condicionada en forma de porcentaje con 2 decimales
print(f"La probabilidad condicionada de extraer en segunda posición una bola azul dado que la primera ha sido roja es del {round(P_condicionada2*100, 2)}%")
```

## <font color='steelblue'> 2. Regla de multiplicación

A partir de la fórmula que utilizamos para calcular la *probabilidad condicionada*, podemos despejar la probabilidad de la *intersección* entre el evento $A$ y $B$ de las siguientes formas:

* $P(A \cap B) = P(B) \cdot P(A|B)$
* $P(A \cap B) = P(A) \cdot P(B|A)$

Estas "nuevas" formas de la fórmula son la **regla de multiplicación** y nos ayudarán en situaciones en las que conocemos la *probabilidad condicionada* de dos sucesos, pero realmente queremos saber la probabilidad de la *intersección*. Esta fórmula se puede interpretar gracias a un tipo de gráfico que se llama **diagrama de árbol**.

<center><img src=https://raw.githubusercontent.com/ia4legos/Statistics/main/images/img_270_20_10.png width="500" height="475"></center>

En la figura anterior, $S$ es nuestro *espacio muestral* y $A$ y $B$ son *sucesos* o *eventos*. La probabilidad en cada punto se obtiene al multiplicar las probabilidades de las ramas que conducen a ese punto. Este tipo de diagrama nos será útil para algunos problemas.

Un ejemplo sencillo para aplicar la *regla de multiplicación* es en el ejemplo del apartado anterior con las bolas. Si sabemos la probabilidad de extraer una bola azul en segundo lugar dado que la primera fue roja y también tenemos la probabilidad de extraer una bola roja, podemos obtener la probabilidad de extraer una bola roja y una azul, en ese orden.

```python
# Calculamos la probabilidad de la intersección multiplicando la probabilidad condicionada por la de extraer una bola roja
P_roja_y_azul = P_condicionada2*P_roja

# Mostramos la probabilidad de la intersección en forma de porcentaje con 2 decimales
print(f"La probabilidad de extraer primero una bola roja y luego una azul, es del {round(P_roja_y_azul*100, 2)}%")
```

Si tenemos tres sucesos $A$, $B$, y $C$ la regla de la multiplicación queda:

$$P(A \cap B \cap C) = P(A) \cdot P(B|A) \cdot P(C|A \cap B)$$

Ppodemos generalizar la *regla de multiplicación* a $n$ sucesos:

$$
P(A_1 \cap A_2 \cap \ldots \cap A_n) = P(A_1) \cdot P(A_2|A_1) \cdot P(A_3|A_2 \cap A_1)  \ldots P(A_n | A_{n-1} \cap A_{n-2} \cap \ldots \cap A_1)
$$

Para aplicar estos nuevos conocimientos a algo más práctico, vemos el siguiente ejemplo:

En una fábrica de móviles se selecciona una muestra de 1000 de ellos para realizar un control de calidad. Se sabe que 7 de ellos están defectuosos. Vamos a extraer al azar 4 de estos móviles, teniendo en cuenta el orden y sin poder volver a cogerlos. ¿Cuál será la probabilidad de que ninguno de estos 4 sea uno de los defectuosos?

Para resolver este problema, vamos a empezar por definir 4 eventos: $A$, que será que el primer móvil no sea defectuoso, $B$, que el segundo móvil no sea defectuoso, $C$, que el tercer móvil no sea defectuoso, y $D$, que el cuarto móvil no sea defectuoso. Teniendo esto, queremos saber la probabilidad de la unión de estos 4 sucesos, es decir, $P(A \cap B \cap C \cap D)$, que sabemos que es igual a $P(A) \cdot P(B|A) \cdot P(C|A \cap B) \cdot P(D|A \cap B \cap C)$. La probabilidad de que el primer móvil extraído no sea defectuoso, es lo mismo que dividir 993 (número de móviles totales menos los defectuosos, 1000-7) entre 1000 (número total de móviles), teniendo así $P(A) = \frac{993}{1000}$. Una vez hemos realizado la primera extracción, nos quedan 999 móviles y 7 siguen siendo defectuosos, por lo que ahora la probabilidad de que el segundo móvil no sea defectuoso, sabiendo que el primero tampoco lo es, será la división de 992 (número de móviles totales, ahora, menos los defectuosos, 999-7) entre 999 (número total de móviles, ahora), teniendo así $P(B|A) = \frac{992}{999}$. Ya hemos seleccionado 2 móviles, por lo que ahora nuestra muestra tienen un tamaño de 998 y siguen estando dentro los 7 defectuosos, así que para saber la probabilidad de que el tercer móvil no sea defectuoso dado que el primero y el segundo tampoco lo son, será la división de 991 (número de móviles totales, ahora, menos los defectuosos, 998-7) entre 998 (número total de móviles, ahora), teniendo así $P(C|A \cap B) = \frac{991}{998}$. Por último, ahora nos queda una muestra de 993 móviles al haber seleccionado ya 3, pero siguen estando dentro los 7 defectuosos, entonces, para saber la probabilidad de que el cuarto móvil no sea defectuoso dado que el primero, el segundo y el tercero tampoco lo son, será la división de 990 (número de móviles totales, ahora, menos los defectuosos, 997-7) entre 997 (número total de móviles, ahora), teniendo así $P(D|A \cap B \cap D) = \frac{990}{997}$. Teniendo todos estos resultados, ya podemos sustituir en nuestra fórmula:

$$
P(A \cap B \cap C \cap D) = P(A) \cdot P(B|A) \cdot P(C|A \cap B) \cdot P(D|A \cap B \cap C) = \frac{993}{1000} \cdot \frac{992}{999} \cdot \frac{991}{998} \cdot \frac{990}{997} = 0.9723
$$

Obtenemos así que, la probabilidad de que al extraer 4 móviles al azar ninguno sea defectuoso, es del 97,23%.

## <font color='steelblue'> 3. Independencia

Se dice que 2 sucesos, $A$ y $B$, son **independientes** si el hecho de que se produzca $A$ no afecta a la probabilidad de que ocurra $B$ o viceversa, es decir, si uno no transmite ninguna información sobre el otro. Algunos ejemplos de *sucesos independientes* son: lanzar una moneda o un dado varias veces. Y algunos *sucesos dependientes* pueden ser: lesionarse en el mismo equipo o la secuenciación de genes en miembros de la misma familia. Al no afectar un suceso al otro, esto lo podemos expresar como $P(B|A) = P(B)$, el suceso $A$ no afecta a $B$, o como $P(A|B) = P(A)$, el suceso $B$ no afecta a $A$. Con la fórmula del apartado anterior, y teniendo $A$ y $B$ como sucesos *independientes*, podemos obtener: $P(A \cap B) = P(A) \cdot P(B|A) = P(A) \cdot P(B)$. Estas 3 fórmulas que hemos puesto significan lo mismo: la independencia entre el evento $A$ y el $B$.

Un primer ejemplo que podemos ver es el lanzamiento de un dado de 6 caras sin trucar. El suceso $A$ corresponde a los números impares ($\{1, 3, 5\}$) y $B$ son los números mayores o iguales que 3 ($\{3, 4, 5, 6\}$). Como nuestro tamaño muestral $|S|$ es 6 ($\{1, 2, 3, 4, 5, 6\}$), tenemos que $P(A) = \frac{3}{6} = \frac{1}{2}$ y que $P(B) = \frac{4}{6} = \frac{2}{3}$. También sabemos que la muestra concordante a la intersección entre $A$ y $B$ es $A \cap B = \{3, 5\}$, con lo que $P(A \cap B) = \frac{2}{6} = \frac{1}{3}$. Si multiplicamos la probabilidad del primer suceso y del segundo: $P(A) \cdot P(B) = \frac{1}{2} \cdot \frac{2}{3} = \frac{1}{3}$. Obtenemos el mismo resultado que antes y corroboramos que $P(A \cap B) = P(A) \cdot P(B)$, por lo que demostramos que el evento $A$ y el $B$ son *independientes*.

Otro ejemplo es si tenemos que el suceso $A$ es que llueva mañana, con una probabilidad $P(A) = \frac{1}{3}$. Y también sabemos que lanzamos una moneda al aire y el suceso $B$ es que salga cara, con una probabilidad $P(B) = \frac{1}{2}$. Queremos calcular la probabilidad de que llueva mañana dado que hemos obtenido una cara al lanzar la moneda, $P(A|B)$. Es fácil de ver que esta es $\frac{1}{3}$, ya que $P(A|B) = P(A)$ al no tener nada que ver el resultado al lanzar una moneda con el tiempo de mañana. No nos importará si se da $B$ o no, la probabilidad de $A$ no cambiará.

Con todo esto, vamos a dar una definición de la independencia más formal:

> Dos sucesos $A$ y $B$ son *independientes* si y solo si $P(A \cap B) = P(A) \cdot P(B)$.

Con esta definición, podemos demostrar la primera fórmula con la que hemos empezado el apartado: $P(B|A) = \frac{P(A \cap B)}{P(A)} = \frac{P(A) \cdot P(B)}{P(A)} = P(B)$. Entonces, si tenemos 2 sucesos *independientes*, $A$ y $B$, y sabemos que $P(A) \neq 0$, entonces $P(B|A) = P(B)$. De la misma manera podemos encontrar que $P(A|B) = P(A)$, si $P(B) \neq 0$. Por lo tanto, podemos decir que $B$ no cambiará nuestra creencia sobre la probabilidad de $A$ o viceversa.

En resumidas cuentas, podemos decir que la **condición de independencia** que deben cumplir dos sucesos para serlo es que, si multiplicamos sus probabilidades, debe dar la probabilidad de la intersección, o bien, que la *probabilidad condicionada* entre ellos sea la misma que la probabilidad original. A veces, la independencia entre 2 sucesos es bastante clara ya que estos no parecen tener ninguna relación física entre sí (como llover y lanzar una moneda). Pero otras veces no es tan fácil saberlo y hay que comprobar si se cumplen la *condición de independencia*, como con el lanzamiento del dado.

## <font color='steelblue'> 4. Ley de probabilidad total

Imaginemos que tenemos un *espacio muestral* $S$ con *particiones*, es decir, sucesos *disjuntos* ($B1$, $B2$,...) cuya unión forma el *espacio muestral*, y añadimos un evento $A$ a este espacio. ¿Cómo podremos calcular la probabilidad de este suceso ($P(A)$)? Este problema es el que afronta la **Ley de probabilidad total**. Antes de ver la solución matemática intentemos entender el problema con unos gráficos.

Consideramos 3 particiones ($B1$, $B2$ y $B3$) de nuestro espacio muestral $S$.

<center><img src=https://raw.githubusercontent.com/ia4legos/Statistics/main/images/img_270_40_20.png width="375" height="250"></center>

A nuestro espacio muestral se añade el evento $A$.

Por lo tanto, al unir las particiones y el evento, nos queda así nuestro espacio muestral.

Ahora tenemos $A1$, $A2$ y $A3$, que son particiones del evento $A$. Estas se obtienen de:

* $A \cap B1$
* $A \cap B2$
* $A \cap B3$

Con estas 3 particiones, podemos obtener la probabilidad total de $A$ si sumamos las probabilidades de cada una de las intersecciones: $P(A) = P(A1) + P(A2) + P(A3)$.

Definimos entonces

> **Ley de probabilidad total.** Si $B_1$, $B_2$, $B_3$, ..., $B_n$ son particiones del espacio muestral $S$, para calcular la probabilidad de cualquier evento $A$ tenemos: $P(A) = \sum_i^n P(A \cap B_i) = \sum_i^n P(A|B_i) \cdot P(B_i)$.

La situación típica en la que se utiliza la *ley de probabilidad total* es cuando queremos hallar la probabilidad de un evento, pero no sabemos encontrarla directamente. Lo que sí sabemos es la probabilidad condicionada de este suceso dadas las particiones del espacio muestral. Sabiendo todo esto ya sobre dicha *ley*, vamos a poner un par de ejemplos.

**Ejemplo 1.** Tenemos tres bolsas de tela y cada una contiene 100 canicas, distribuidas de la siguiente manera:

* *Bolsa 1:* 70 rojas y 30 azules.
* *Bolsa 2:* 63 rojas y 37 azules.
* *Bolsa 3:* 26 rojas y 74 azules.

Se escoge una de las tres bolsas al azar y luego se saca una canica de la bolsa seleccionada, también al azar. ¿Cuál será la probabilidad de que la canica extraída sea roja? Primero de todo debemos tener en cuenta que, en este problema, nuestro espacio muestral tiene 3 particiones: cada una de las bolsas de tela. Esto es así porque son sucesos *disyuntos* (solo podemos seleccionar una) y además, su unión forma el espacio muestral total ($P(Bolsa1 \cup Bolsa2 \cup Bolsa3) = 1$), ya que una de las bolsas será elegida sí o sí. Por lo tanto, cada una de las probabilidades será $\frac{1}{3}$. Luego, vamos a considerar que $R$ es el suceso de escoger una canica roja y tenemos que $Bolsa1$ es el evento de seleccionar la primera bolsa, $Bolsa2$ el evento de seleccionar la segunda bolsa y $Bolsa3$ el evento de seleccionar la tercera bolsa. Una vez definido esto, sabemos que:

* $P(R|Bolsa1) = 0.70$
* $P(R|Bolsa2) = 0.63$
* $P(R|Bolsa3) = 0.26$

Con todos estos datos, podemos aplicar la *ley de probabilidad total*:

$$
P(R) = \sum_i^3 P(R|Bolsa_i) \cdot P(Bolsa_i) = P(R|Bolsa1) \cdot P(Bolsa1) + P(R|Bolsa2) \cdot P(Bolsa2) + P(R|Bolsa3) \cdot P(Bolsa3) = 0.70 \cdot \frac{1}{3} + 0.63 \cdot \frac{1}{3} + 0.26 \cdot \frac{1}{3} = 0.53
$$

Obtenemos de esta manera que tenemos una probabilidad del 53% de extraer una canica roja. Este problema vamos a resolverlo también con un *diagrama de árbol* y con **Python**. Empezamos mostrando el diagrama:

<center><img src=https://raw.githubusercontent.com/ia4legos/Statistics/main/images/img_270_40_50.png width="550" height="475"></center>

El evento $A$ indica que la canica extraída es azul. Entonces, si sumamos las 3 probabilidades que mostramos, obtenemos el resultado de $P(A)$ que hemos calculado antes. Para hacerlo en **Python**, crearemos 6 variables: una para cada una de las probabilidades seleccionar alguna de las 3 bolsas y otras 3 para las probabilidades condicionadas de seleccionar una canica roja. Luego aplicaremos la *ley de probabilidad total*.

```python
# Probabilidad de seleccionar cualquiera de las 3 bolsas
P_Bolsa1 = P_Bolsa2 = P_Bolsa3 = 1/3

# Probabilidades condicionadas
P_Roja_dado_Bolsa1 = 0.7
P_Roja_dado_Bolsa2 = 0.63
P_Roja_dado_Bolsa3 = 0.26

# Ley de probabilidad total
P_Roja = P_Roja_dado_Bolsa1*P_Bolsa1 + P_Roja_dado_Bolsa2*P_Bolsa2 + P_Roja_dado_Bolsa3*P_Bolsa3

# Mostramos la probabilidad
print(f"La probabilidad de extraer una canica roja es del {P_Roja*100}%")
```

**Ejemplo 2.** Disponemos de unos datos sobre un estudio que se le hizo a un grupo de personas de 50 años en un hospital. A estos individuos se les clasifica en 3 grupos según su riesgo de ictus: bajo, medio o alto. De todos los que estaban en el estudio, se clasificó a un 60% como de riesgo bajo, un 30% como riesgo medio y un 10% con un riesgo alto. Se les estuvo haciendo un seguimiento durante 5 años y se constataron los siguientes datos:

* De los individuos con *riesgo bajo* de ictus, el 1% sufrió un ictus.
* De los individuos con *riesgo medio* de ictus, el 5% sufrió un ictus.
* De los individuos con *riesgo alto* de ictus, el 9% sufrió un ictus.

Si seleccionamos al azar una persona de 50 años, ¿cuál será la probabilidad de que sufra un ictus? Para resolver este problema vamos a utilizar un *diagrama de árbol*, pero primero vamos a identificar las particiones y el evento. Nuestro espacio muestral $S$ tiene 3 particiones: $Riesgo_{Bajo}$, $Riesgo_{Medio}$ y $Riesgo_{Alto}$, además, el suceso $I$ haber sufrido un ictus, mientras que $\bar{I}$ será no haberlo sufrido. Mostremos ahora el *diagrama* para este problema:

<center><img src=https://raw.githubusercontent.com/ia4legos/Statistics/main/images/img_270_40_60.png width="600" height="475"></center>

Con los tres resultados que obtenemos ($P(I \cap Riesgo_{Bajo})$, $P(I \cap Riesgo_{Medio})$ y $P(I \cap Riesgo_{Alto})$) aplicamos la *ley de probabilidad total* para hallar la probabilidad de nuestro evento $I$:

$$
P(I) = \sum_i^3 P(I \cap Riesgo_i) = P(I \cap Riesgo_{Bajo}) + P(I \cap Riesgo_{Medio}) + P(I \cap Riesgo_{Alto}) = 0.006 + 0.015 + 0.009 = 0.03

Por lo que podemos concluir que una persona que tiene 50 años tiene un riesgo del 3% de sufrir un ictus.

## <font color='steelblue'> 5. Teorema de Bayes

Ya hemos visto fórmulas de tipos de probabilidad, reglas, leyes y demostrar si dos o más sucesos son independientes o no. Nos falta por ver uno de los resultados más útiles de la probabilidad condicionada: el **Teorema de Bayes**. Supongamos que en un problema sabemos cuánto vale $P(A|B)$, pero nosotros a nosotros nos interesa la probabilidad $P(B|A)$. ¿Cómo podemos averiguarlo? El teorema de Bayes nos da respuesta a esta pregunta:

> **Teorema de Bayes.**
* Para 2 sucesos cualesquiera $A$ y $B$, donde $P(A) \neq 0$, tenemos $$P(B|A) = \frac{P(A|B) \cdot P(B)}{P(A)}.$$
* Si $B_1$, $B_2$, ..., $B_n$ son las $n$ particiones del espacio muestral $S$ y $A$ es cualquier suceso con $P(A) \neq 0$, tenemos $$P(B_i|A) = \frac{P(A|B_i) \cdot P(B_i)}{\sum_j^nP(A|B_j) \cdot P(B_j)}.$$

Podemos volver a los 2 ejemplos que hemos puesto en el apartado anterior para aplicar este *teorema*.

**Ejemplo 1**. Sabemos que se ha extraído una canica roja, ¿cuál será la probabilidad de que haya salido de la tercera bolsa? En este caso tenemos el valor de $P(A)$, pero imaginemos que no lo tenemos. Como ya tenemos el *diagrama de árbol*, vamos a resolver el problema con sus datos aplicando la fórmula del *teorema* y con **Python**. Tenemos que $P(Bolsa1) = P(Bolsa2) = P(Bolsa3) = \frac{1}{3}$, $P(R|Bolsa1) = 0.70$, $P(R|Bolsa2) = 0.63$ y $P(R|Bolsa3) = 0.26$. Si aplicamos la fórmula del *Teorema de Bayes* para un espacio muestral con 3 particiones:

$$
P(Bolsa3|R) = \frac{P(R|Bolsa3) \cdot P(Bolsa3)}{\sum_j^3P(R|Bolsa_j) \cdot P(Bolsa_j)} = \frac{0.26 \cdot \frac{1}{3}}{0.70 \cdot \frac{1}{3} + 0.63 \cdot \frac{1}{3} + 0.26 \cdot \frac{1}{3}} = 0.1635
$$

Obtenemos de esta manera que la probabilidad de haber extraído una canica roja haya sido de la tercera bolsa es del $16,35\%$. Si nos fijamos, este valor es menor que $33,33\%$ ($\frac{1}{3}$), que es la probabilidad de selección de la tercera bolsa. Esto es así porque esta bolsa tiene menos canicas rojas que el resto de las bolsas, por lo que, si sabemos que se ha escogido una canica roja, será menos probable que haya sido de la tercera bolsa.

Este resultado también lo podemos obteniendo con código de **Python**.

```python
# Teorema de Bayes
P_Bolsa3_dado_Roja = (P_Roja_dado_Bolsa3*P_Bolsa3)/(P_Roja_dado_Bolsa1*P_Bolsa1 + P_Roja_dado_Bolsa2*P_Bolsa2 + P_Roja_dado_Bolsa3*P_Bolsa3)

# Mostramos la probabilidad
print(f"La probabilidad de haber estraido de la tercera bolsa, dado que es una canica roja es del {round(P_Bolsa3_dado_Roja*100, 2)}%")
```

**Ejemplo 2.** Con los datos del estudio sobre el riesgo de sufrir un ictus en personas de 50 años, queremos saber la probabilidad de pertenecer al grupo con las características de riesgo medio de sufrirlo si sabemos que no ha tenido ningún ictus. En este caso vamos a utilizar el dato de $P(I) = 3\%$. Nuestro suceso aquí será $\bar{I}$, es decir, no sufrir un ictus, por lo que su probabilidad será el complementario de la anterior $P(\bar{I}) = 1 - P(I) = 1 - 0.03 = 0.97$ . Si recordamos de nuestro *diagrama de árbol* tenemos el valor de $P(\bar{I}|Riesgo_{Medio})$ es $0.95$. Con estos datos ya podemos sustituir con el *Teorema de Bayes*:

$$
P(Riesgo_{Medio}|\bar{I}) = \frac{P(\bar{I}|Riesgo_{Medio}) \cdot P(Riesgo_{Medio})}{P(\bar{I})} = \frac{0.95 \cdot 0.3}{0.97} = 0.2938
$$

El grupo de riesgo con mayor número de sujetos era el de riesgo bajo, con un 60% de ellos, mientras que los de riesgo medio de sufrir un ictus representan el 30% del total. Por esto es normal que, si sabemos que una persona no ha sufrido un ictus, haya una probabilidad del $29,38\%$ de que pertenezca al grupo de riesgo medio.
