### **EJ001**

Para el experimento aleatorio asociado a lanzar al mismo tiempo dos dados (equilibrados), escribe los 36 resultados posibles que pueden darse. Con los datos obtenidos responde a las siguientes cuestiones:

*   ¿Cuál es la probabilidad de que salga un 1 en el primer dado y un 2 en el segundo dado?
*   ¿Cuál es la probabilidad de que salga un 1 en el primer dado y un número par en el segundo dado?
*   ¿Cuál es la probabilidad de que salga un 1 en el primer dado y un número impar en el segundo dado?

```python
import itertools

d1 = {"1", "2", "3", "4", "5", "6"}
d2 = {"1", "2", "3", "4", "5", "6"}

soluciones = []
for i in itertools.product(d1, d2):
  soluciones.append(i)

print("Los 36 resultados posibles son:")
print(soluciones)
print(f"\nTamaño muestral: {len(soluciones)}")
```


*   **Apartado 1: ¿Cuál es la probabilidad de que salga un 1 en el primer dado y un 2 en el segundo dado?**

```python
suma_ap1 = 0
for resultado in soluciones:
  # Verificar si el primer dado es '1' y el segundo dado es '2'
  if resultado[0] == '1' and resultado[1] == '2':
    suma_ap1 += 1

probabilidad_ap1 = suma_ap1 / len(soluciones)
print(f"Probabilidad (1 en primer dado y 2 en segundo dado): {probabilidad_ap1:.4f}")
```

*   **Apartado 2: ¿Cuál es la probabilidad de que salga un 1 en el primer dado y un número par en el segundo dado?**

```python
suma_ap2 = 0
# Números pares en un dado
pares = {'2', '4', '6'}

for resultado in soluciones:
  # Verificar si el primer dado es '1' y el segundo dado es par
  if resultado[0] == '1' and resultado[1] in pares:
    suma_ap2 += 1

probabilidad_ap2 = suma_ap2 / len(soluciones)
print(f"Probabilidad (1 en primer dado y número par en segundo dado): {probabilidad_ap2:.4f}")
```

*   **Apartado 3: ¿Cuál es la probabilidad de que salga un 1 en el primer dado y un número impar en el segundo dado?**

```python
suma_ap3 = 0
# Números impares en un dado
impares = {'1', '3', '5'}

for resultado in soluciones:
  # Verificar si el primer dado es '1' y el segundo dado es impar
  if resultado[0] == '1' and resultado[1] in impares:
    suma_ap3 += 1

probabilidad_ap3 = suma_ap3 / len(soluciones)
print(f"Probabilidad (1 en primer dado y número impar en segundo dado): {probabilidad_ap3:.4f}")
```
